## 🛍️ Case Study 1: E-commerce Order Analytics System
### Difficulty: ⭐⭐ Intermediate

---

### 💼 Business Context

**Company:** Mid-size e-commerce retailer (10K orders/day, $50M annual revenue)

**Problem Statement:**
> "We need a data warehouse to answer business questions like: What's our daily revenue? Which products are top sellers? What's the average order value by customer segment? Currently, this data is scattered across operational databases and takes hours to compile manually."

**Business Requirements:**
1. **Daily KPI Dashboard** - Revenue, orders, avg order value (AOV)
2. **Product Performance** - Best/worst sellers by category
3. **Customer Segmentation** - RFM analysis (Recency, Frequency, Monetary)
4. **Operational Metrics** - Order fulfillment time, cancellation rate
5. **Executive Reports** - Month-over-month growth, quarterly trends

**Data Volume:**
- 10K orders/day (≈ 3.6M orders/year)
- 50K products, 500K customers
- 3 years of historical data to load initially

**Latency Requirements:**
- Daily batch updates (overnight refresh acceptable)
- Queries must complete in < 10 seconds

---

### 📦 Data Sources

**Source Systems:**

1. **PostgreSQL (Transactional DB)** - Orders, order_items, products
2. **MongoDB (User Service)** - Customer profiles, preferences
3. **CSV Files (Legacy)** - Historical product catalog

**Source Schema (Simplified):**

```sql
-- Source: PostgreSQL orders table
orders: order_id, customer_id, order_date, status, shipping_cost, tax_amount, total_amount

-- Source: PostgreSQL order_items table  
order_items: order_item_id, order_id, product_id, quantity, unit_price, discount

-- Source: PostgreSQL products table
products: product_id, product_name, category, subcategory, cost_price, list_price

-- Source: MongoDB customers collection
customers: {customer_id, name, email, city, state, country, signup_date, segment}
```

---

### 🏛️ Target Schema Design (Star Schema)

**Design Decision:** Star schema for fast analytical queries

**Why Star Schema?**
✅ Fast JOINs (fact to dimensions only, no dimension-to-dimension)
✅ Simple for BI tools (Tableau, Power BI)
✅ Denormalized dimensions = fewer JOINs
✅ Query patterns are well-known (aggregations, filtering, grouping)

**Trade-off:** More storage (denormalized), but storage is cheap in cloud.

---

````python
%sql
-- ==============================================
-- DIMENSION TABLES
-- ==============================================

-- Dimension: Date
CREATE TABLE workspace.default.dim_date (
  date_key INT PRIMARY KEY,  -- 20240101
  date_value DATE,
  year INT,
  quarter INT,
  month INT,
  month_name STRING,
  week_of_year INT,
  day_of_month INT,
  day_of_week INT,
  day_name STRING,
  is_weekend BOOLEAN,
  is_holiday BOOLEAN,
  fiscal_year INT,
  fiscal_quarter INT
);

-- Dimension: Customer (SCD Type 1 - overwrite)
CREATE TABLE workspace.default.dim_customer (
  customer_key INT PRIMARY KEY,  -- Surrogate key
  customer_id STRING,            -- Natural key from source
  customer_name STRING,
  email STRING,
  city STRING,
  state STRING,
  country STRING,
  customer_segment STRING,       -- 'Enterprise', 'SMB', 'Consumer'
  signup_date DATE,
  is_active BOOLEAN,
  last_updated_at TIMESTAMP
);

-- Dimension: Product (SCD Type 2 - track history)
CREATE TABLE workspace.default.dim_product (
  product_key INT PRIMARY KEY,   -- Surrogate key
  product_id STRING,             -- Natural key
  product_name STRING,
  category STRING,
  subcategory STRING,
  cost_price DECIMAL(10,2),
  list_price DECIMAL(10,2),
  is_active BOOLEAN,
  
  -- SCD Type 2 fields
  effective_date DATE,
  expiration_date DATE,
  is_current BOOLEAN,
  
  last_updated_at TIMESTAMP
);

-- ==============================================
-- FACT TABLE
-- ==============================================

CREATE TABLE workspace.default.fact_orders (
  -- Surrogate key for fact
  order_fact_key BIGINT PRIMARY KEY,
  
  -- Foreign keys to dimensions
  date_key INT,
  customer_key INT,
  
  -- Degenerate dimensions (high-cardinality IDs)
  order_id STRING,
  order_item_id STRING,
  
  -- Foreign key to product (SCD Type 2 aware)
  product_key INT,
  
  -- Additive measures
  quantity INT,
  unit_price DECIMAL(10,2),
  discount_amount DECIMAL(10,2),
  line_total DECIMAL(10,2),      -- quantity * unit_price - discount
  shipping_cost DECIMAL(10,2),
  tax_amount DECIMAL(10,2),
  
  -- Semi-additive measures  
  cost_price DECIMAL(10,2),      -- For profit calc
  
  -- Non-additive facts (descriptive)
  order_status STRING,           -- 'pending', 'shipped', 'delivered', 'cancelled'
  
  -- Audit columns
  created_at TIMESTAMP,
  updated_at TIMESTAMP,
  
  -- Constraints
  CONSTRAINT fk_date FOREIGN KEY (date_key) REFERENCES dim_date(date_key),
  CONSTRAINT fk_customer FOREIGN KEY (customer_key) REFERENCES dim_customer(customer_key),
  CONSTRAINT fk_product FOREIGN KEY (product_key) REFERENCES dim_product(product_key)
);

-- Indexes for performance
CREATE INDEX idx_fact_orders_date ON workspace.default.fact_orders(date_key);
CREATE INDEX idx_fact_orders_customer ON workspace.default.fact_orders(customer_key);
CREATE INDEX idx_fact_orders_product ON workspace.default.fact_orders(product_key);
CREATE INDEX idx_fact_orders_status ON workspace.default.fact_orders(order_status);
````

### 🔄 ETL Strategy & Data Flow

#### **Architecture Overview:**

```
Source Systems          Staging           Data Warehouse         BI Tools
┌───────────────┐   ┌─────────────┐   ┌────────────────┐   ┌─────────────┐
│ PostgreSQL      │   │ Bronze Layer │   │ dim_date       │   │ Tableau     │
│ (orders, items) │──>│ (raw copy)   │──>│ dim_customer   │──>│ Power BI    │
│                 │   │              │   │ dim_product    │   │ Looker      │
└───────────────┘   │              │   │ fact_orders    │   └─────────────┘
┌───────────────┐   │ Silver Layer │   └────────────────┘
│ MongoDB         │   │ (cleaned)    │
│ (customers)     │──>│              │
└───────────────┘   └─────────────┘
┌───────────────┐   ┌─────────────┐
│ CSV Files       │   │ Gold Layer   │
│ (products)      │──>│ (business)  │
└───────────────┘   └─────────────┘
```

#### **ETL Approach: Incremental Load**

**Why Incremental?**
- ✅ Faster: Only process changed/new data
- ✅ Lower cost: Less compute and network
- ✅ Easier testing: Can reprocess specific dates

**Change Data Capture (CDC) Strategy:**

1. **Orders/Items:** Track by `updated_at` timestamp
2. **Customers:** Upsert based on `customer_id` (SCD Type 1)
3. **Products:** SCD Type 2 for price changes

**Pseudo-code:**
```python
# Daily ETL job (runs at 2 AM)
max_order_date = get_max_date_from_dwh()

# Extract: Pull orders from yesterday
new_orders = extract_orders(where=f"order_date > '{max_order_date}'")

# Transform: JOIN with dimensions, calc metrics
transformed = transform_orders(new_orders)

# Load: Append to fact table
load_to_fact_table(transformed, mode='append')

# Refresh aggregates
refresh_materialized_views()
```

---

````python
%sql
-- ==============================================
-- INCREMENTAL ETL EXAMPLE
-- ==============================================

-- Step 1: Get max date already loaded
CREATE OR REPLACE TEMP VIEW max_loaded_date AS
SELECT COALESCE(MAX(d.date_value), '1900-01-01') AS last_loaded_date
FROM workspace.default.fact_orders f
JOIN workspace.default.dim_date d ON f.date_key = d.date_key;

-- Step 2: Extract new/changed orders from source (simulated)
CREATE OR REPLACE TEMP VIEW source_orders_incremental AS
SELECT 
  'ORD005' AS order_id,
  'ITEM006' AS order_item_id,
  'C002' AS customer_id,
  'P003' AS product_id,
  DATE('2024-03-15') AS order_date,
  2 AS quantity,
  299.99 AS unit_price,
  20.00 AS discount,
  'pending' AS order_status,
  10.00 AS shipping_cost,
  45.60 AS tax_amount,
  150.00 AS cost_price;

-- Step 3: Transform - JOIN with dimensions to get surrogate keys
CREATE OR REPLACE TEMP VIEW transformed_orders AS
SELECT 
  ROW_NUMBER() OVER (ORDER BY s.order_id) + 
    (SELECT COALESCE(MAX(order_fact_key), 0) FROM workspace.default.fact_orders) AS order_fact_key,
  d.date_key,
  c.customer_key,
  s.order_id,
  s.order_item_id,
  p.product_key,
  s.quantity,
  s.unit_price,
  s.discount,
  s.quantity * s.unit_price - s.discount AS line_total,
  s.shipping_cost,
  s.tax_amount,
  s.cost_price,
  s.order_status,
  CURRENT_TIMESTAMP() AS created_at,
  CURRENT_TIMESTAMP() AS updated_at
FROM source_orders_incremental s
JOIN workspace.default.dim_date d 
  ON d.date_value = s.order_date
JOIN workspace.default.dim_customer c 
  ON c.customer_id = s.customer_id
JOIN workspace.default.dim_product p 
  ON p.product_id = s.product_id 
  AND p.is_current = TRUE  -- SCD Type 2: Join to current record
WHERE s.order_date > (SELECT last_loaded_date FROM max_loaded_date);

-- Step 4: Load - Insert new facts
INSERT INTO workspace.default.fact_orders
SELECT * FROM transformed_orders;

SELECT 
  COUNT(*) AS rows_inserted,
  MIN(date_key) AS first_date,
  MAX(date_key) AS last_date
FROM transformed_orders;
````

### 📊 Business Questions & SQL Queries

**Query 1: Daily Revenue Dashboard**
```sql
SELECT 
  d.date_value,
  d.day_name,
  COUNT(DISTINCT f.order_id) AS total_orders,
  SUM(f.line_total) AS gross_revenue,
  SUM(f.line_total - (f.cost_price * f.quantity)) AS gross_profit,
  ROUND(AVG(f.line_total), 2) AS avg_order_value,
  COUNT(DISTINCT f.customer_key) AS unique_customers
FROM fact_orders f
JOIN dim_date d ON f.date_key = d.date_key
WHERE f.order_status != 'cancelled'
  AND d.date_value >= CURRENT_DATE - INTERVAL 30 DAY
GROUP BY d.date_value, d.day_name
ORDER BY d.date_value DESC;
```

**Query 2: Top Products by Category**
```sql
SELECT 
  p.category,
  p.product_name,
  SUM(f.quantity) AS units_sold,
  SUM(f.line_total) AS revenue,
  SUM(f.line_total - (f.cost_price * f.quantity)) AS profit,
  COUNT(DISTINCT f.customer_key) AS unique_buyers
FROM fact_orders f
JOIN dim_product p ON f.product_key = p.product_key
WHERE f.order_status = 'delivered'
GROUP BY p.category, p.product_name
QUALIFY ROW_NUMBER() OVER (PARTITION BY p.category ORDER BY SUM(f.line_total) DESC) <= 5
ORDER BY p.category, revenue DESC;
```

**Query 3: Customer RFM Segmentation**
```sql
WITH customer_metrics AS (
  SELECT 
    c.customer_key,
    c.customer_name,
    c.customer_segment,
    MAX(d.date_value) AS last_order_date,
    DATEDIFF(CURRENT_DATE, MAX(d.date_value)) AS recency_days,
    COUNT(DISTINCT f.order_id) AS frequency,
    SUM(f.line_total) AS monetary_value
  FROM fact_orders f
  JOIN dim_customer c ON f.customer_key = c.customer_key
  JOIN dim_date d ON f.date_key = d.date_key
  WHERE f.order_status != 'cancelled'
  GROUP BY c.customer_key, c.customer_name, c.customer_segment
),
rfm_scores AS (
  SELECT *,
    NTILE(5) OVER (ORDER BY recency_days ASC) AS r_score,
    NTILE(5) OVER (ORDER BY frequency DESC) AS f_score,
    NTILE(5) OVER (ORDER BY monetary_value DESC) AS m_score
  FROM customer_metrics
)
SELECT 
  customer_name,
  customer_segment,
  recency_days,
  frequency,
  monetary_value,
  r_score,
  f_score,
  m_score,
  CASE 
    WHEN r_score >= 4 AND f_score >= 4 THEN 'Champions'
    WHEN r_score >= 3 AND f_score >= 3 THEN 'Loyal Customers'
    WHEN r_score >= 4 AND f_score <= 2 THEN 'Promising'
    WHEN r_score <= 2 AND f_score >= 4 THEN 'At Risk'
    WHEN r_score <= 2 AND f_score <= 2 THEN 'Lost'
    ELSE 'Potential'
  END AS rfm_segment
FROM rfm_scores
ORDER BY monetary_value DESC;

### 🚦 Fraud Detection Rules & Scoring

#### **Rule-Based Checks (Hard Rules)**

```sql
-- Rule 1: Impossible Travel
WITH geo_check AS (
  SELECT 
    t.*,
    v.last_latitude,
    v.last_longitude,
    v.last_txn_timestamp,
    -- Haversine distance formula (simplified)
    6371 * ACOS(
      COS(RADIANS(t.latitude)) * 
      COS(RADIANS(v.last_latitude)) * 
      COS(RADIANS(v.last_longitude) - RADIANS(t.longitude)) +
      SIN(RADIANS(t.latitude)) * 
      SIN(RADIANS(v.last_latitude))
    ) AS distance_km,
    TIMESTAMPDIFF(SECOND, v.last_txn_timestamp, t.transaction_timestamp) / 3600.0 AS hours_elapsed
  FROM transactions_stream t
  JOIN user_velocity v ON t.user_id = v.user_id
)
SELECT *,
  CASE 
    WHEN hours_elapsed > 0 AND (distance_km / hours_elapsed) > 800 
    THEN 'IMPOSSIBLE_TRAVEL'  -- Faster than airplane
    ELSE NULL
  END AS fraud_flag
FROM geo_check;
```

**Rule 2: Velocity Breach**
```sql
SELECT *,
  CASE 
    WHEN txn_count_1min > 5 THEN 'VELOCITY_1MIN'
    WHEN txn_count_5min > 10 THEN 'VELOCITY_5MIN'
    WHEN amount_sum_30min > 10000 THEN 'VELOCITY_AMOUNT'
    ELSE NULL
  END AS velocity_flag
FROM transactions_stream t
JOIN user_velocity v ON t.user_id = v.user_id;
```

**Rule 3: High-Risk Merchant**
```sql
SELECT t.*,
  CASE 
    WHEN m.fraud_rate > 0.10 THEN 'HIGH_RISK_MERCHANT'
    WHEN m.merchant_age_days < 30 THEN 'NEW_MERCHANT'
    ELSE NULL
  END AS merchant_flag
FROM transactions_stream t
JOIN merchant_risk m ON t.merchant_id = m.merchant_id;

#### **Feature Engineering for ML Model**

```sql
WITH features AS (
  SELECT 
    t.transaction_id,
    
    -- Transaction features
    t.amount,
    HOUR(t.transaction_timestamp) AS txn_hour,
    DAYOFWEEK(t.transaction_timestamp) AS txn_day_of_week,
    
    -- Velocity features
    v.txn_count_1min,
    v.txn_count_5min,
    v.txn_count_30min,
    v.amount_sum_30min,
    v.unique_merchants_1hour,
    
    -- Behavioral anomaly (Z-score)
    (t.amount - b.avg_transaction_amount) / NULLIF(b.stddev_transaction_amount, 0) AS amount_zscore,
    ABS(HOUR(t.transaction_timestamp) - b.typical_hour_of_day) AS hour_deviation,
    
    -- Geo features
    CASE WHEN t.country != v.last_country THEN 1 ELSE 0 END AS is_new_country,
    CASE WHEN t.device_id != v.last_device_id THEN 1 ELSE 0 END AS is_new_device,
    
    -- Merchant risk
    m.fraud_rate AS merchant_fraud_rate,
    m.merchant_age_days,
    
    -- Historical context
    b.total_transactions AS user_txn_count,
    b.historical_fraud_count AS user_fraud_count,
    DATEDIFF(t.transaction_timestamp, b.first_seen_date) AS user_age_days
    
  FROM transactions_stream t
  JOIN user_velocity v ON t.user_id = v.user_id
  JOIN user_baseline b ON t.user_id = b.user_id
  JOIN merchant_risk m ON t.merchant_id = m.merchant_id
)
SELECT * FROM features;
```

#### **Scoring Logic (Combining Rules + ML)**

```sql
WITH rule_scores AS (
  SELECT 
    transaction_id,
    -- Hard rule flags
    CASE WHEN impossible_travel THEN 100 ELSE 0 END +
    CASE WHEN velocity_breach THEN 80 ELSE 0 END +
    CASE WHEN high_risk_merchant THEN 60 ELSE 0 END AS rule_score
  FROM fraud_checks
),
ml_scores AS (
  SELECT 
    transaction_id,
    ml_fraud_probability * 100 AS ml_score  -- 0-100 scale
  FROM ml_predictions
),
combined_scores AS (
  SELECT 
    t.transaction_id,
    GREATEST(
      COALESCE(r.rule_score, 0),
      COALESCE(m.ml_score, 0)
    ) AS final_fraud_score,
    CASE 
      WHEN COALESCE(r.rule_score, 0) >= 100 OR COALESCE(m.ml_score, 0) >= 95 THEN 'critical'
      WHEN COALESCE(r.rule_score, 0) >= 60 OR COALESCE(m.ml_score, 0) >= 80 THEN 'high'
      WHEN COALESCE(m.ml_score, 0) >= 50 THEN 'medium'
      ELSE 'low'
    END AS risk_tier,
    CASE 
      WHEN COALESCE(r.rule_score, 0) >= 100 OR COALESCE(m.ml_score, 0) >= 95 THEN 'blocked'
      WHEN COALESCE(r.rule_score, 0) >= 60 OR COALESCE(m.ml_score, 0) >= 80 THEN 'review'
      ELSE 'approved'
    END AS decision
  FROM transactions_stream t
  LEFT JOIN rule_scores r ON t.transaction_id = r.transaction_id
  LEFT JOIN ml_scores m ON t.transaction_id = m.transaction_id
)
SELECT * FROM combined_scores;
```

> "Write a query to find duplicate customer records based on email address. Show all duplicates, not just one representative. This is a common data quality check."

### ✅ Answer 8: Duplicate Detection Strategies

#### **Three Approaches:**

**1. Self JOIN (Classic)**
```sql
SELECT c1.*
FROM Customers c1
JOIN Customers c2 
  ON c1.email = c2.email
  AND c1.id != c2.id;
```

**2. GROUP BY + HAVING (More Efficient)**
```sql
SELECT email, COUNT(*) AS duplicate_count
FROM Customers
GROUP BY email
HAVING COUNT(*) > 1;
```

**3. Window Function (Best for large data)**
```sql
SELECT *
FROM (
  SELECT *,
    ROW_NUMBER() OVER (PARTITION BY email ORDER BY id) AS rn
  FROM Customers
)
WHERE rn > 1;  -- Keep only duplicates
```

#### **Performance Comparison:**

| Method | Performance | Shows All Dupes | Delete-Ready |
|--------|-------------|-----------------|-------------|
| Self JOIN | ⭐⭐ Slow | ✅ Yes | ❌ No |
| GROUP BY | ⭐⭐⭐⭐ Fast | ❌ No (summary) | ❌ No |
| Window Function | ⭐⭐⭐⭐⭐ Fastest | ✅ Yes | ✅ Yes |

#### **Complete Deduplication Pattern:**

```sql
-- Step 1: Find duplicates with row numbers
WITH ranked AS (

    ROW_NUMBER() OVER (
      PARTITION BY email 
      ORDER BY created_date DESC, id DESC  -- Keep newest
    ) AS rn
  FROM Customers
)

DELETE FROM Customers
WHERE id IN (

-- Good: Start with filtered orders

WHERE o.order_date >= '2024-01-01'
JOIN Customers c ON o.customer_id = c.id

-- Bad: Filter after all joins

JOIN B ...
```

## 📌 Section 5: GROUP BY Fundamentals (4 Questions)

GROUP BY is essential for aggregations. Understanding execution order and common patterns is critical.

### ❓ Question 14: SQL Execution Order

**Fundamental Interview Question:**
> "Explain the order in which SQL clauses are executed. Why does this matter for GROUP BY queries?"

### ✅ Answer 14: SQL Execution Order

#### **Logical Execution Order:**

```
1. FROM       -- Tables and JOINs
2. WHERE      -- Row filtering (BEFORE aggregation)
3. GROUP BY   -- Group rows
4. HAVING     -- Group filtering (AFTER aggregation)
5. SELECT     -- Column selection and computation
6. ORDER BY   -- Sort results
7. LIMIT      -- Limit results
```

#### **Why This Matters:**

**1. WHERE vs HAVING:**
```sql
-- WHERE filters rows BEFORE grouping (faster!)
WHERE salary > 50000

-- HAVING filters groups AFTER aggregation
HAVING AVG(salary) > 50000
```

**2. Column References:**
```sql
-- This FAILS:
SELECT department, AVG(salary) AS avg_sal
FROM employees
WHERE avg_sal > 50000;  -- ❌ avg_sal doesn't exist yet!

-- This WORKS:
SELECT department, AVG(salary) AS avg_sal
FROM employees
GROUP BY department
HAVING AVG(salary) > 50000;  -- ✅ Aggregation available in HAVING
```

**3. GROUP BY Requirements:**
* Non-aggregated columns in SELECT MUST be in GROUP BY
* Aggregate functions (SUM, COUNT, AVG) allowed in SELECT without GROUP BY

**4. Performance Optimization:**
```sql
-- Bad: Filter after aggregation
SELECT department, COUNT(*)
FROM employees
GROUP BY department
HAVING department = 'Sales';  -- ❌ Slow!

-- Good: Filter before aggregation
SELECT department, COUNT(*)
FROM employees
WHERE department = 'Sales'    -- ✅ Fast!
GROUP BY department;
```

#### **Common Interview Trap:**
> "Why can't I use a SELECT alias in WHERE?"

**Answer:** WHERE executes BEFORE SELECT, so aliases don't exist yet!

````python
%sql
-- Demo: WHERE vs HAVING

-- Method 1: Filter with WHERE (BEFORE aggregation) - FASTER
SELECT 
  Department,
  COUNT(*) AS EmployeeCount,
  AVG(Salary) AS AvgSalary
FROM workspace.default.Employees
WHERE Salary > 60000  -- Filter individual rows first
GROUP BY Department
ORDER BY AvgSalary DESC;
````

### ❓ Question 16: COUNT(*) vs COUNT(column) vs COUNT(DISTINCT column)

**Common Interview Question:**
> "Explain the difference between COUNT(*), COUNT(column), and COUNT(DISTINCT column). When would you use each?"

### ✅ Answer 16: Understanding COUNT Variants

#### **Three Types:**

**1. COUNT(*) - Counts ALL rows**
```sql
SELECT COUNT(*) FROM employees;

```

**2. COUNT(column) - Counts NON-NULL values**
```sql
SELECT COUNT(manager_id) FROM employees;
-- Returns: Number of employees with a manager (excludes NULLs)
```

**3. COUNT(DISTINCT column) - Counts unique NON-NULL values**
```sql
SELECT COUNT(DISTINCT department) FROM employees;
-- Returns: Number of unique departments
```

#### **Example with Data:**

```
employees:
id | name    | manager_id | department
1  | Alice   | NULL       | Sales
2  | Bob     | 1          | Sales
3  | Charlie | 1          | IT
4  | David   | 2          | NULL

COUNT(*):                     4  (all rows)
COUNT(manager_id):            2  (Bob, Charlie have managers)
COUNT(department):            3  (David has NULL)
COUNT(DISTINCT department):   2  (Sales, IT)
```

#### **Performance Comparison:**

```
Dataset: 100M rows

COUNT(*):           Fast    (row count)
COUNT(column):      Medium  (check NULLs)
COUNT(DISTINCT):    Slow    (deduplication overhead)
```

#### **Common Patterns:**

**Find missing data:**
```sql
SELECT

  COUNT(email) AS has_email,
  COUNT(*) - COUNT(email) AS missing_email
FROM customers;
```

**Uniqueness check:**
```sql
SELECT 
  COUNT(*) AS total,
  COUNT(DISTINCT email) AS unique_emails
FROM customers;
-- If total != unique_emails, you have duplicates!
```

#### **Important:**
⚠️ **COUNT(*) is fastest** - Use when you need total rows
⚠️ **COUNT(DISTINCT) is expensive** - Consider approximate functions for huge data

### ❓ Question 17: GROUP BY with ROLLUP/CUBE

**Advanced Question:**
> "Explain ROLLUP and CUBE extensions to GROUP BY. Provide a use case where ROLLUP would be beneficial."

### ✅ Answer 17: ROLLUP and CUBE

#### **ROLLUP - Hierarchical Subtotals**

Creates subtotals at each level of grouping hierarchy:

```sql
SELECT region, department, SUM(sales)
FROM sales_data
GROUP BY ROLLUP(region, department);
```

**Returns:**
* Detail: Each (region, department) combination
* Subtotal: Each region (all departments)
* Grand total: All regions and departments

#### **CUBE - All Possible Combinations**

```sql
GROUP BY CUBE(region, department);
```

**Returns all combinations:**
* (region, department)
* (region, NULL) - By region
* (NULL, department) - By department
* (NULL, NULL) - Grand total

#### **Use Cases:**

**ROLLUP for hierarchical reporting:**
```sql
-- Sales by Year → Quarter → Month
SELECT 
  year, quarter, month,
  SUM(sales) AS total_sales
FROM sales
GROUP BY ROLLUP(year, quarter, month);
```

**Result:**
```
Year | Quarter | Month | Sales
2024 | Q1      | Jan   | 100
2024 | Q1      | Feb   | 150
2024 | Q1      | NULL  | 250  (Q1 subtotal)
2024 | NULL    | NULL  | 250  (2024 subtotal)
NULL | NULL    | NULL  | 250  (Grand total)
```

#### **Identify Subtotal Rows:**
```sql
SELECT 
  COALESCE(region, 'ALL REGIONS') AS region,
  COALESCE(department, 'ALL DEPARTMENTS') AS department,
  SUM(sales) AS total_sales
FROM sales_data
GROUP BY ROLLUP(region, department);
```

## 📌 Section 6: HAVING vs WHERE (2 Questions)

Understanding when to filter with WHERE vs HAVING is a common interview topic.

### ❓ Question 18: HAVING vs WHERE - When to Use Each

**Very Common Interview Question:**
> "What's the difference between WHERE and HAVING? Why can't you use aggregate functions in WHERE? Provide examples of each."

### ✅ Answer 18: WHERE vs HAVING

#### **Key Differences:**

| Aspect | WHERE | HAVING |
|--------|-------|--------|
| **When** | BEFORE GROUP BY | AFTER GROUP BY |
| **Filters** | Individual rows | Grouped results |
| **Can use aggregates** | ❌ NO | ✅ YES |
| **Performance** | ✅ Faster (fewer rows) | 🐌 Slower (more rows processed) |

#### **Execution Order:**
```
FROM → WHERE → GROUP BY → HAVING → SELECT → ORDER BY
```

#### **WHERE Examples:**
```sql
-- Filter rows BEFORE aggregation
SELECT department, AVG(salary)
FROM employees
WHERE salary > 50000      -- Filter individual salaries
GROUP BY department;
```

#### **HAVING Examples:**
```sql
-- Filter groups AFTER aggregation
SELECT department, AVG(salary) AS avg_sal
FROM employees
GROUP BY department
HAVING AVG(salary) > 70000;  -- Filter department averages
```

#### **Why No Aggregates in WHERE?**

**WHERE executes BEFORE GROUP BY**, so groups don't exist yet!

```sql
-- This FAILS:
SELECT department, COUNT(*)
FROM employees
WHERE COUNT(*) > 5        -- ❌ Error! No groups yet
GROUP BY department;

-- This WORKS:
SELECT department, COUNT(*)
FROM employees
GROUP BY department
HAVING COUNT(*) > 5;      -- ✅ Groups exist now
```

#### **Combining Both:**
```sql
-- Best practice: Filter early with WHERE, then filter groups with HAVING
SELECT department, AVG(salary) AS avg_sal
FROM employees
WHERE hire_date >= '2020-01-01'  -- WHERE: Filter rows
GROUP BY department
HAVING AVG(salary) > 60000;      -- HAVING: Filter groups
```

#### **Performance Rule:**
⚡ **Filter with WHERE whenever possible** - Reduces data before aggregation!

```sql
-- Bad: Process all departments, then filter
GROUP BY department
HAVING department = 'Sales';  -- 🐌 Slow

-- Good: Filter early
WHERE department = 'Sales'    -- ✅ Fast
GROUP BY department;
```

````python
%sql
-- Demo: WHERE vs HAVING

-- Query 1: Using WHERE (filter rows before grouping)
SELECT 
  Department,
  COUNT(*) AS EmployeeCount,
  AVG(Salary) AS AvgSalary
FROM workspace.default.Employees
WHERE Salary > 50000  -- Filter: Only high salaries
GROUP BY Department
ORDER BY AvgSalary DESC;
````

````python
%sql
-- Query 2: Using HAVING (filter groups after aggregation)
SELECT 
  Department,
  COUNT(*) AS EmployeeCount,
  AVG(Salary) AS AvgSalary
FROM workspace.default.Employees
GROUP BY Department
HAVING COUNT(*) >= 2  -- Filter: Only departments with 2+ employees
ORDER BY AvgSalary DESC;
````

```sql
SUM(CASE WHEN salary < 50000 THEN 1 ELSE 0 END) AS low,
SUM(CASE WHEN salary BETWEEN 50000 AND 80000 THEN 1 ELSE 0 END) AS mid,
SUM(CASE WHEN salary > 80000 THEN 1 ELSE 0 END) AS high
```

**4. Pivoting data:**
```sql
SELECT 
  product_id,
  SUM(CASE WHEN region = 'North' THEN sales ELSE 0 END) AS north_sales,
  SUM(CASE WHEN region = 'South' THEN sales ELSE 0 END) AS south_sales,
  SUM(CASE WHEN region = 'East' THEN sales ELSE 0 END) AS east_sales
FROM sales
GROUP BY product_id;
```

#### **Why Use CASE in Aggregations?**

✅ Single query instead of multiple queries
✅ Consistent aggregation (same grouping)
✅ Better performance (one scan)
✅ Easy to calculate percentages across categories

#### **Alternative: FILTER Clause (SQL Standard)**
```sql
-- Some databases support FILTER (Databricks does not yet)
COUNT(*) FILTER (WHERE salary > 70000) AS high_earners
```

````python
%sql
-- Conditional aggregation: Categorize employees by salary
SELECT 
  Department,
  COUNT(*) AS TotalEmployees,
  SUM(CASE WHEN Salary > 70000 THEN 1 ELSE 0 END) AS HighEarners,
  SUM(CASE WHEN Salary <= 70000 THEN 1 ELSE 0 END) AS StandardEarners,
  ROUND(100.0 * SUM(CASE WHEN Salary > 70000 THEN 1 ELSE 0 END) / COUNT(*), 2) AS PctHighEarners,
  ROUND(AVG(CASE WHEN Salary > 70000 THEN Salary END), 2) AS AvgHighSalary,
  ROUND(AVG(CASE WHEN Salary <= 70000 THEN Salary END), 2) AS AvgStandardSalary
FROM workspace.default.Employees
GROUP BY Department
ORDER BY TotalEmployees DESC;
````

### ❓ Question 21: Running Totals and Cumulative Sums

**Interview Question:**
> "Calculate running total of sales by date. Show each date's sales and the cumulative total. Compare window function vs self-join approaches."

### ✅ Answer 21: Running Totals (Cumulative Sums)

#### **Method 1: Window Function (BEST)**
```sql
SELECT 
  date,
  sales,
  SUM(sales) OVER (ORDER BY date) AS running_total,
  AVG(sales) OVER (ORDER BY date ROWS BETWEEN 2 PRECEDING AND CURRENT ROW) AS moving_avg_3
FROM daily_sales
ORDER BY date;
```

**Pros:** ✅ Fast, ✅ Clean, ✅ Flexible

#### **Method 2: Self-JOIN (Legacy)**
```sql
SELECT 
  s1.date,
  s1.sales,
  SUM(s2.sales) AS running_total
FROM sales s1
JOIN sales s2 ON s2.date <= s1.date
GROUP BY s1.date, s1.sales
ORDER BY s1.date;
```

**Pros:** Works without window functions  
**Cons:** 🐌 O(N²) performance, very slow

#### **Window Frame Options:**

```sql
-- Cumulative from start
SUM(sales) OVER (ORDER BY date)

-- Last 7 days
SUM(sales) OVER (ORDER BY date ROWS BETWEEN 6 PRECEDING AND CURRENT ROW)

-- Centered 3-day average
AVG(sales) OVER (ORDER BY date ROWS BETWEEN 1 PRECEDING AND 1 FOLLOWING)

-- Partition by category
SUM(sales) OVER (PARTITION BY category ORDER BY date)
```

#### **Performance:**
```
Dataset: 1M rows

Window Function:  3.2 seconds  ⭐⭐⭐⭐⭐
Self-JOIN:        12 minutes   ⭐
```

#### **Common Use Cases:**
* Running balance in financial reports
* Cumulative sales/revenue
* Moving averages for trend analysis
* Year-to-date (YTD) metrics

````python
%sql
-- Running total of sales by region
SELECT 
  Region,
  SaleDate,
  Amount,
  SUM(Amount) OVER (PARTITION BY Region ORDER BY SaleDate) AS RunningTotal,
  AVG(Amount) OVER (PARTITION BY Region ORDER BY SaleDate ROWS BETWEEN 2 PRECEDING AND CURRENT ROW) AS MovingAvg3
FROM workspace.default.SalesData
ORDER BY Region, SaleDate
LIMIT 15;
````

### ❓ Question 22: Percentiles and Distribution Analysis

**Interview Question:**
> "Find the 25th, 50th (median), 75th, and 90th percentile salaries by department. Explain PERCENTILE_CONT vs PERCENTILE_DISC."

### ✅ Answer 22: Percentile Analysis

#### **PERCENTILE_CONT (Continuous)**
Interpolates between values:

```sql
SELECT 
  department,
  PERCENTILE_CONT(0.25) WITHIN GROUP (ORDER BY salary) AS p25,
  PERCENTILE_CONT(0.50) WITHIN GROUP (ORDER BY salary) AS median,
  PERCENTILE_CONT(0.75) WITHIN GROUP (ORDER BY salary) AS p75,
  PERCENTILE_CONT(0.90) WITHIN GROUP (ORDER BY salary) AS p90
FROM employees
GROUP BY department;
```

#### **PERCENTILE_DISC (Discrete)**
Returns actual value from dataset:

```sql
PERCENTILE_DISC(0.50) WITHIN GROUP (ORDER BY salary)
```

#### **Difference:**

```
Salaries: [10, 20, 30, 40]

PERCENTILE_CONT(0.50):  25  (average of 20 and 30)
PERCENTILE_DISC(0.50):  20  (actual value from data)
```

#### **Alternative: PERCENT_RANK()**

```sql
SELECT *,
  PERCENT_RANK() OVER (PARTITION BY department ORDER BY salary) AS percentile
FROM employees;
-- Returns 0 to 1 showing relative position
```

#### **Use Cases:**
* Salary bands (P25, P50, P75)
* Performance distribution
* Identifying outliers (> P95 or < P5)
* Market analysis (median pricing)

#### **NTILE for Equal Buckets:**
```sql
SELECT *,
  NTILE(4) OVER (ORDER BY salary) AS quartile
FROM employees;
-- Divides into 4 equal-sized buckets
```

````python
%sql
-- Salary percentiles by department
SELECT 
  Department,
  COUNT(*) AS EmployeeCount,
  MIN(Salary) AS MinSalary,
  PERCENTILE_CONT(0.25) WITHIN GROUP (ORDER BY Salary) AS P25_Salary,
  PERCENTILE_CONT(0.50) WITHIN GROUP (ORDER BY Salary) AS Median_Salary,
  PERCENTILE_CONT(0.75) WITHIN GROUP (ORDER BY Salary) AS P75_Salary,
  MAX(Salary) AS MaxSalary
FROM workspace.default.Employees
GROUP BY Department
ORDER BY Median_Salary DESC;
````

## 📌 Section 8: Aggregate Functions with NULL Handling (3 Questions)

NULLs behave differently across aggregate functions. Understanding this is critical for data quality.

### ❓ Question 23: How Do Aggregate Functions Handle NULLs?

**Fundamental Interview Question:**
> "Explain how NULL values are handled by SUM, AVG, COUNT, MIN, MAX. Provide examples where NULL handling causes unexpected results."

### ✅ Answer 23: NULLs in Aggregate Functions

#### **General Rule:**
Most aggregate functions **IGNORE NULLs** (except COUNT(*)).

#### **Behavior by Function:**

```
Data: [10, 20, NULL, 30]

COUNT(*):        4  (counts all rows)
COUNT(column):   3  (ignores NULLs)
SUM(column):    60  (10+20+30, skips NULL)
AVG(column):    20  (60/3, NOT 60/4!)
MIN(column):    10  (ignores NULL)
MAX(column):    30  (ignores NULL)
```

#### **Critical AVG Pitfall:**

⚠️ **AVG ignores NULLs** - This can give misleading results!

```sql
-- Grades: [90, 80, NULL, 70]

AVG(grade):  80  (240/3, ignores NULL)
-- But if NULL means "not submitted" (grade = 0), correct average is 60!

COUNT(*):    4
COUNT(grade): 3  -- Missing assignment not counted!
```

**Solution: COALESCE**
```sql
AVG(COALESCE(grade, 0))  -- Treats NULL as 0
```

#### **Empty Groups Return NULL:**

```sql
SELECT department, SUM(sales)
FROM sales
WHERE 1=0  -- No rows
GROUP BY department;
-- Returns: NULL (not 0!)
```

**Solution: COALESCE in SELECT**
```sql
COALESCE(SUM(sales), 0) AS total_sales
```

#### **NULL in GROUP BY:**

NULLs are treated as a single group:

```sql
-- manager_id: [1, 1, NULL, NULL, 2]

SELECT manager_id, COUNT(*)
FROM employees
GROUP BY manager_id;

-- Result:
1,    2
NULL, 2  -- All NULLs grouped together
2,    1
```

#### **Best Practices:**

✅ Use **COUNT(*)** for row counts
✅ Use **COUNT(column)** to check for NULLs
✅ Use **COALESCE** when NULL should be treated as a value
✅ Check for NULL groups in results

#### **Data Quality Checks:**
```sql
SELECT 
  COUNT(*) AS total_rows,
  COUNT(email) AS has_email,
  COUNT(*) - COUNT(email) AS missing_email,
  100.0 * COUNT(email) / COUNT(*) AS pct_complete
FROM customers;
```

````python
%sql
-- Demo: How aggregates handle NULLs
SELECT 
  COUNT(*) AS TotalRows,
  COUNT(ManagerID) AS HasManager,
  COUNT(*) - COUNT(ManagerID) AS NoManager,
  SUM(Salary) AS TotalSalary,
  AVG(Salary) AS AvgSalary,
  MIN(ManagerID) AS MinManagerID,
  MAX(ManagerID) AS MaxManagerID
FROM workspace.default.Employees;
````

### ❓ Question 24: COALESCE vs IFNULL vs NVL

**Interview Question:**
> "Explain the difference between COALESCE, IFNULL, and NVL. When would you use each? Can COALESCE accept more than 2 arguments?"

### ✅ Answer 24: NULL Replacement Functions

#### **Three Functions:**

**1. COALESCE (SQL Standard)**
```sql
COALESCE(value1, value2, value3, ...)
-- Returns first non-NULL value
```

**2. IFNULL (MySQL, Databricks)**
```sql
IFNULL(value, replacement)
-- Returns value if not NULL, else replacement
```

**3. NVL (Oracle)**
```sql
NVL(value, replacement)
-- Same as IFNULL, Oracle syntax
```

#### **Key Differences:**

| Function | Arguments | Standard | Databases |
|----------|-----------|----------|----------|
| COALESCE | 2+ (unlimited) | ✅ SQL Standard | All |
| IFNULL | Exactly 2 | ❌ Proprietary | MySQL, Databricks |
| NVL | Exactly 2 | ❌ Proprietary | Oracle |

#### **Examples:**

**Single fallback:**
```sql
-- All equivalent:
COALESCE(email, 'no-email@example.com')
IFNULL(email, 'no-email@example.com')
NVL(email, 'no-email@example.com')
```

**Multiple fallbacks (COALESCE only):**
```sql
COALESCE(work_email, personal_email, 'no-email@example.com')
-- Returns first non-NULL from the list
```

#### **Common Use Cases:**

**1. Replace NULL in calculations:**
```sql
SELECT 
  first_name,
  COALESCE(bonus, 0) AS bonus,
  salary + COALESCE(bonus, 0) AS total_comp
FROM employees;
```

**2. Cascading fallbacks:**
```sql
COALESCE(
  mobile_phone,
  work_phone,
  home_phone,
  'No phone on file'
) AS contact_number
```

**3. Aggregation safety:**
```sql
SUM(COALESCE(sales, 0)) AS total_sales
-- Ensures NULL doesn't break sum
```

**4. Display formatting:**
```sql
COALESCE(manager_name, 'No Manager') AS manager
```

#### **Performance:**

All three have similar performance. Use COALESCE for portability.

#### **Interview Tip:**
⚡ **COALESCE is preferred** - it's SQL standard and more flexible!

````python
%sql
-- Demo: COALESCE for NULL handling
SELECT 
  FirstName,
  LastName,
  ManagerID,
  COALESCE(ManagerID, 0) AS ManagerID_Safe,
  CASE 
    WHEN ManagerID IS NULL THEN 'Top Level'
    ELSE 'Has Manager'
  END AS ManagerStatus,
  Salary,
  COALESCE(Salary * 0.1, 0) AS Bonus
FROM workspace.default.Employees
ORDER BY ManagerID NULLS FIRST;
````

### ❓ Question 25: NULL-Safe Equality and Comparisons

**Advanced Interview Question:**
> "Explain why `column = NULL` doesn't work. How do you compare NULLs? What is NULL-safe equality (<=>)?"

### ✅ Answer 25: NULL Comparison Rules

#### **Fundamental Rule:**
⚠️ **NULL = NULL is NOT TRUE!**

NULL represents "unknown", so comparing unknowns is undefined.

#### **Common Mistakes:**

```sql
-- WRONG: This returns NO rows, even if column IS NULL!
WHERE column = NULL

-- WRONG: This also fails
WHERE column != NULL

-- RIGHT: Use IS NULL
WHERE column IS NULL
WHERE column IS NOT NULL
```

#### **NULL in Expressions:**

```sql
NULL + 10     = NULL
NULL * 5      = NULL
NULL > 10     = NULL (not TRUE or FALSE!)
'Hello' || NULL = NULL
```

**Any operation with NULL returns NULL!**

#### **NULL in Boolean Logic:**

```sql
TRUE AND NULL   = NULL
FALSE AND NULL  = FALSE  (short-circuit)
TRUE OR NULL    = TRUE   (short-circuit)
FALSE OR NULL   = NULL
NOT NULL        = NULL
```

#### **NULL-Safe Equality (<=>):**

```sql
-- Standard comparison:
NULL = NULL     → NULL (not TRUE!)

-- NULL-safe equality (MySQL, Databricks):
NULL <=> NULL   → TRUE
10 <=> 10       → TRUE
10 <=> NULL     → FALSE
```

**Equivalent using IS:**
```sql
-- Instead of a <=> b:
(a = b) OR (a IS NULL AND b IS NULL)
```

#### **JOIN with NULLs:**

```sql
-- This FAILS to match NULL keys:
ON t1.manager_id = t2.manager_id

-- Use NULL-safe equality:
ON t1.manager_id <=> t2.manager_id
```

#### **IN with NULLs:**

```sql
WHERE id IN (1, 2, NULL)
-- id=3 returns FALSE (not NULL!)
-- IN converts to: id=1 OR id=2 OR id=NULL
```

#### **Best Practices:**

✅ **Always use IS NULL / IS NOT NULL for NULL checks**
✅ **Use COALESCE to handle NULLs in comparisons**
✅ **Use <=> for NULL-safe equality** (if supported)
✅ **Document whether NULLs are expected in data**

#### **Interview Trick Question:**

> "What does `WHERE NOT (status = 'active')` return?"

**Answer:** Excludes both 'active' AND NULL!

```sql
-- If you want to include NULLs:
WHERE status != 'active' OR status IS NULL
```

````python
%sql
-- Demo: NULL comparison behaviors

-- Comparison 1: Regular equality (NULLs don't match)
SELECT 
  'Test 1: Regular Equality' AS Test,
  FirstName,
  ManagerID
FROM workspace.default.Employees
WHERE ManagerID = NULL  -- This returns 0 rows!
LIMIT 5;
````

````python
%sql
-- Comparison 2: Correct NULL check
SELECT 
  'Test 2: IS NULL' AS Test,
  FirstName,
  LastName,
  ManagerID,
  Department
FROM workspace.default.Employees
WHERE ManagerID IS NULL  -- Correct way to find NULLs
ORDER BY Department;
````
