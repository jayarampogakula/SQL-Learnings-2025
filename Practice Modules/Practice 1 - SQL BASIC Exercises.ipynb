## Exercise 3: High Salary Employees 🟢

**Problem:** Find all employees earning more than $70,000.

**Expected Output:** Employees with Salary > 70000

**Difficulty:** Easy

````python
%sql
-- Write your query here

````

````python
%sql
-- SOLUTION
SELECT FirstName, LastName, Department, Salary
FROM workspace.default.Employees
WHERE Salary > 70000
ORDER BY Salary DESC;
````
