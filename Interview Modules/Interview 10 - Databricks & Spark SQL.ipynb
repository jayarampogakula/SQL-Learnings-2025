## 📌 Section 1: Delta Lake Fundamentals (5 Questions)

Delta Lake brings ACID transactions, time travel, and schema evolution to data lakes.

### ❓ Question 1: Explain Delta Lake vs Parquet

**Fundamental Interview Question:**
> "What's the difference between a Delta table and a Parquet table? When would you use each? What are the trade-offs?"

### ✅ Answer 1: Delta Lake vs Parquet

#### **Quick Answer**

**Parquet:** Columnar file format (storage layer only)
**Delta Lake:** Transaction layer on top of Parquet files

---

#### **Detailed Comparison**

| Feature | Parquet Table | Delta Lake Table |
|---------|---------------|------------------|
| **File Format** | Parquet files only | Parquet + Delta transaction log |
| **ACID Transactions** | ❌ No | ✅ Yes (serializable isolation) |
| **Time Travel** | ❌ No | ✅ Yes (version history) |
| **Schema Evolution** | ❌ Manual | ✅ Automatic (mergeSchema) |
| **UPSERT/MERGE** | ❌ Not supported | ✅ Native MERGE command |
| **Concurrent Writes** | ❌ Overwrites conflict | ✅ Optimistic concurrency control |
| **Streaming** | ❌ Batch only | ✅ Streaming + batch unified |
| **DELETE/UPDATE** | ❌ Rewrite entire partition | ✅ Row-level deletes |
| **Small File Problem** | ❌ Manual compaction | ✅ Auto-optimize available |
| **Data Skipping** | ✅ Min/max stats | ✅ Enhanced stats + Z-order |
| **Performance** | Fast reads | Fast reads + ACID overhead |
| **Storage Cost** | Lower (no logs) | Slightly higher (transaction logs) |

---

#### **How Delta Lake Works**

```
Delta Table Structure:
/path/to/table/
  ├── _delta_log/          ← Transaction log (JSON)
  │   ├── 00000000000000000000.json
  │   ├── 00000000000000000001.json
  │   └── 00000000000000000010.checkpoint.parquet
  └── part-00000-*.parquet ← Actual data files
```

**Transaction Log:**
- Records every transaction (add/remove files)
- Enables ACID guarantees
- Supports time travel queries
- Checkpoints every 10 commits for performance

---

#### **Example: ACID Guarantees**

**Parquet (No ACID):**
```python
# Writer 1 starts
df1.write.mode("append").parquet("/path/to/table")

# Writer 2 starts simultaneously
df2.write.mode("append").parquet("/path/to/table")  # ⚠️ Race condition!

# Result: Files may be lost or duplicated
```

**Delta Lake (ACID):**
```python
# Writer 1 starts
df1.write.format("delta").mode("append").save("/path/to/table")

# Writer 2 starts simultaneously
df2.write.format("delta").mode("append").save("/path/to/table")

# ✅ Both writes succeed, no data loss
# ✅ Transaction log ensures serializability
```

---

#### **When to Use Each**

**Use Parquet When:**
- Read-only datasets (no updates)
- Long-term archival storage
- Interoperability with non-Delta tools
- Minimizing storage costs (no transaction logs)

**Use Delta Lake When:**
- Concurrent read/write workloads
- Need UPSERT/MERGE operations
- Streaming data ingestion
- Schema evolution required
- Audit/compliance needs (time travel)
- Production data pipelines (ACID guarantees)

---

#### **Migration: Parquet → Delta**

```sql
-- Convert existing Parquet table to Delta
CONVERT TO DELTA parquet.`/path/to/parquet_table`;

-- Or create Delta table from Parquet
CREATE TABLE delta_table
USING DELTA
AS SELECT * FROM parquet.`/path/to/parquet_table`;
```

---

#### **Performance Trade-offs**

**Delta Overhead:**
- ~5-10% storage overhead (transaction logs)
- Small write latency (log updates)
- Read performance: **Same or faster** (better stats)

**Delta Benefits:**
- Data skipping with enhanced stats
- Z-ordering for co-location
- Adaptive Query Execution optimization

---

#### **Interview Follow-Up Questions**

**Q: "Can you read Delta tables with non-Databricks tools?"**

**A:** "Yes, with Delta Standalone library (Java/Scala) or using Delta Lake connectors for:
- Apache Spark (open-source)
- Presto/Trino

- Pandas (via deltalake-python)

But advanced features (time travel, MERGE) require Delta-aware engines."

---

**Q: "What happens if you delete _delta_log folder?"**

**A:** "Table becomes unreadable! The transaction log is the source of truth. Without it:
- Can't determine which Parquet files are current
- Lose all version history
- Lose schema information

Always back up _delta_log with table backups."

---

**Q: "How does Delta Lake handle small files?"**

**A:** "Delta Lake provides:
1. **Auto Optimize** - Automatically compacts small files on write
2. **OPTIMIZE command** - Manual compaction
3. **Optimized Writes** - Creates right-sized files (128MB default)

```sql
-- Enable auto-optimize
ALTER TABLE my_table SET TBLPROPERTIES (
  'delta.autoOptimize.optimizeWrite' = 'true',
  'delta.autoOptimize.autoCompact' = 'true'
);
```

````python
%sql
-- Demo: Delta Lake vs Parquet Comparison

-- Create sample data
CREATE OR REPLACE TABLE workspace.default.sample_data AS
SELECT 
  id,
  CAST(RAND() * 100 AS INT) AS category,
  RAND() * 1000 AS amount,
  CURRENT_DATE() - INTERVAL CAST(RAND() * 365 AS INT) DAY AS date
FROM RANGE(1000000);

-- Write as Parquet
CREATE OR REPLACE TABLE workspace.default.parquet_table
USING PARQUET
AS SELECT * FROM workspace.default.sample_data;

-- Write as Delta
CREATE OR REPLACE TABLE workspace.default.delta_table
USING DELTA
AS SELECT * FROM workspace.default.sample_data;

-- Compare storage size
DESCRIBE DETAIL workspace.default.parquet_table;
DESCRIBE DETAIL workspace.default.delta_table;

-- Time Travel (only works with Delta)
SELECT COUNT(*) FROM workspace.default.delta_table VERSION AS OF 0;
-- SELECT COUNT(*) FROM workspace.default.parquet_table VERSION AS OF 0;  -- ❌ Error!

-- ACID Test: Concurrent appends
INSERT INTO workspace.default.delta_table SELECT * FROM workspace.default.sample_data LIMIT 10;
INSERT INTO workspace.default.delta_table SELECT * FROM workspace.default.sample_data LIMIT 10;
-- ✅ Both succeed with Delta

-- Show transaction history
DESCRIBE HISTORY workspace.default.delta_table;
````

### ❓ Question 2: Delta Lake Time Travel

**Production Scenario Interview:**
> "A data engineer accidentally deleted 1M rows from a production Delta table 2 hours ago. How would you recover the data? What are the limitations of time travel?"

### ✅ Answer 2: Time Travel & Data Recovery

#### **What is Time Travel?**

Delta Lake maintains a **version history** of every transaction:
- Each write creates a new version (monotonically increasing)
- Transaction log stores metadata for all versions
- Old data files remain until VACUUM

---

#### **Time Travel Syntax**

```sql
-- Query by version number
SELECT * FROM my_table VERSION AS OF 42;

-- Query by timestamp
SELECT * FROM my_table TIMESTAMP AS OF '2024-01-15 10:30:00';

-- Using @v syntax
SELECT * FROM my_table@v42;
```

---

#### **Recovery Scenario: Restore Deleted Rows**

**Problem:** Accidentally deleted 1M rows 2 hours ago

**Solution 1: RESTORE Command (Databricks)**

```sql
-- Check current state
SELECT COUNT(*) FROM production.sales;  -- 9M rows (should be 10M)

-- View history to find the issue
DESCRIBE HISTORY production.sales;
/*
version | timestamp           | operation | operationMetrics
15      | 2024-01-15 14:00    | DELETE    | numDeletedRows=1000000
14      | 2024-01-15 12:00    | WRITE     | numOutputRows=10000000
*/

-- RESTORE to version before the delete
RESTORE TABLE production.sales TO VERSION AS OF 14;

-- Verify recovery
SELECT COUNT(*) FROM production.sales;  -- ✅ 10M rows
```

**Solution 2: Insert from Time Travel Query**

```sql
-- Find missing rows by comparing versions
INSERT INTO production.sales
SELECT * FROM production.sales VERSION AS OF 14
EXCEPT
SELECT * FROM production.sales;  -- Current version

-- Verify
SELECT COUNT(*) FROM production.sales;  -- ✅ 10M rows restored
```

**Solution 3: Create Backup Table**

```sql
-- Create backup of good state
CREATE TABLE production.sales_backup AS
SELECT * FROM production.sales VERSION AS OF 14;

-- Swap tables (advanced)
ALTER TABLE production.sales RENAME TO production.sales_corrupt;
ALTER TABLE production.sales_backup RENAME TO production.sales;
```

---

#### **Time Travel Limitations**

**1. Retention Period (Default: 30 days)**

```sql
-- Check retention setting
DESCRIBE DETAIL production.sales;
-- Shows: deletedFileRetentionDuration = 7 days (default)

-- Extend retention
ALTER TABLE production.sales SET TBLPROPERTIES (
  'delta.deletedFileRetentionDuration' = 'interval 90 days'
);
```

**Problem:** After VACUUM, old versions are gone!

```sql
-- VACUUM removes old files (default: 7 days)
VACUUM production.sales RETAIN 168 HOURS;  -- Keep 7 days

-- After VACUUM, time travel before retention fails
SELECT * FROM production.sales VERSION AS OF 5;  -- ❌ Error!
```

**2. Storage Cost**

Time travel keeps old data files:
- Deleted rows still consume storage
- Multiple versions = multiple copies of data
- Balance retention vs cost

**3. Performance**

Querying old versions may be slow:
- Old files not optimized (small files)
- No Z-ordering on historical data

---

#### **Best Practices**

**1. Disable Auto-VACUUM for Critical Tables**

```sql
ALTER TABLE production.critical_table SET TBLPROPERTIES (
  'delta.deletedFileRetentionDuration' = 'interval 365 days',
  'delta.logRetentionDuration' = 'interval 365 days'
);
```

**2. Create Savepoints Before Risky Operations**

```sql
-- Before bulk delete/update
CREATE TABLE production.sales_snapshot_20240115 SHALLOW CLONE production.sales;

-- Perform risky operation
DELETE FROM production.sales WHERE ...;

-- Rollback if needed
RESTORE TABLE production.sales FROM sales_snapshot_20240115;
```

**3. Monitor Version Count**

```sql
-- Check number of versions
SELECT COUNT(*) AS version_count
FROM (DESCRIBE HISTORY production.sales);

-- Alert if versions > 1000 (indicates no VACUUM)
```

---

#### **Advanced: Version Metadata**

```sql
-- Detailed version info
DESCRIBE HISTORY production.sales;

-- Find version at specific time
SELECT version, timestamp, operation
FROM (DESCRIBE HISTORY production.sales)
WHERE timestamp <= '2024-01-15 10:00:00'
ORDER BY version DESC
LIMIT 1;

-- Diff two versions
SELECT 'added' AS change_type, *
FROM production.sales VERSION AS OF 15
EXCEPT
SELECT 'added', * FROM production.sales VERSION AS OF 14

UNION ALL

SELECT 'removed' AS change_type, *
FROM production.sales VERSION AS OF 14
EXCEPT
SELECT 'removed', * FROM production.sales VERSION AS OF 15;
```

---

#### **Interview Follow-Up**

**Q: "Can you time travel on a table that's been VACUUMed?"**

**A:** "Only within the retention period. If you VACUUM with RETAIN 7 DAYS, versions older than 7 days are permanently deleted. Always:
1. Set retention > VACUUM retention
2. Test VACUUM on dev before prod
3. Create SHALLOW CLONEs before risky operations"

````python
%sql
-- Demo: Time Travel & Data Recovery

-- Create sample table
CREATE OR REPLACE TABLE workspace.default.transactions (
  transaction_id INT,
  amount DECIMAL(10,2),
  date DATE
);

-- Version 0: Insert initial data
INSERT INTO workspace.default.transactions VALUES
  (1, 100.00, '2024-01-01'),
  (2, 200.00, '2024-01-02'),
  (3, 300.00, '2024-01-03');

-- Version 1: Add more data
INSERT INTO workspace.default.transactions VALUES
  (4, 400.00, '2024-01-04'),
  (5, 500.00, '2024-01-05');

-- Version 2: Accidental delete!
DELETE FROM workspace.default.transactions WHERE transaction_id >= 4;

-- Check current state
SELECT * FROM workspace.default.transactions;  -- Only 3 rows!

-- View version history
DESCRIBE HISTORY workspace.default.transactions;

-- Time travel to version before delete
SELECT * FROM workspace.default.transactions VERSION AS OF 1;  -- ✅ 5 rows

-- Restore deleted rows
INSERT INTO workspace.default.transactions
SELECT * FROM workspace.default.transactions VERSION AS OF 1
WHERE transaction_id >= 4;

-- Verify recovery
SELECT * FROM workspace.default.transactions ORDER BY transaction_id;

-- Alternative: Full table restore
-- RESTORE TABLE workspace.default.transactions TO VERSION AS OF 1;
````

### ❓ Question 3: Delta MERGE for Upserts

**CDC/Streaming Interview Question:**
> "You're ingesting CDC (Change Data Capture) events from a transactional database into a Delta table. Each event has an operation (INSERT/UPDATE/DELETE). How do you apply these changes efficiently using MERGE?"

### ✅ Answer 3: Delta Lake MERGE (Upserts)

#### **MERGE Syntax Overview**

```sql
MERGE INTO target_table AS target
USING source_table AS source
ON target.id = source.id

WHEN MATCHED AND source.operation = 'DELETE' THEN
  DELETE

WHEN MATCHED AND source.operation = 'UPDATE' THEN
  UPDATE SET
    target.column1 = source.column1,
    target.updated_at = source.updated_at

WHEN NOT MATCHED AND source.operation = 'INSERT' THEN
  INSERT (id, column1, created_at)
  VALUES (source.id, source.column1, source.created_at);
```

---

#### **CDC Use Case: Full Example**

**Scenario:** Syncing customer data from MySQL to Delta Lake

**Source CDC Events:**
```
operation | customer_id | name    | email
INSERT    | 1           | Alice   | alice@email.com
INSERT    | 2           | Bob     | bob@email.com
UPDATE    | 1           | Alice S | alice@email.com
DELETE    | 2           | Bob     | bob@email.com
```

**Target Delta Table:**
```sql
CREATE TABLE production.customers (
  customer_id INT,
  name STRING,
  email STRING,
  updated_at TIMESTAMP,
  is_deleted BOOLEAN DEFAULT FALSE
);
```

**MERGE Query:**

```sql
MERGE INTO production.customers AS target
USING (
  -- Deduplicate CDC events (keep latest per key)
  SELECT customer_id, name, email, operation,
    ROW_NUMBER() OVER (PARTITION BY customer_id ORDER BY event_timestamp DESC) AS rn
  FROM staging.cdc_events
  WHERE event_date = CURRENT_DATE()
) AS source
ON target.customer_id = source.customer_id
  AND source.rn = 1  -- Only process latest event per customer

-- Handle DELETE
WHEN MATCHED AND source.operation = 'DELETE' THEN
  UPDATE SET 
    target.is_deleted = TRUE,
    target.updated_at = CURRENT_TIMESTAMP()

-- Handle UPDATE
WHEN MATCHED AND source.operation = 'UPDATE' THEN
  UPDATE SET
    target.name = source.name,
    target.email = source.email,
    target.updated_at = CURRENT_TIMESTAMP(),
    target.is_deleted = FALSE

-- Handle INSERT
WHEN NOT MATCHED AND source.operation = 'INSERT' THEN
  INSERT (customer_id, name, email, updated_at, is_deleted)
  VALUES (source.customer_id, source.name, source.email, CURRENT_TIMESTAMP(), FALSE);
```

---

#### **MERGE Performance Optimization**

**1. Partition Pruning**

```sql
-- Add partition hint to avoid full table scan
MERGE INTO production.customers AS target
USING source
ON target.customer_id = source.customer_id
  AND target.partition_date >= CURRENT_DATE() - INTERVAL 7 DAYS  -- ✅ Prune partitions!
...
```

**2. Z-Ordering on Join Keys**

```sql
-- Optimize target table for MERGE performance
OPTIMIZE production.customers ZORDER BY (customer_id);

-- MERGE will be 3-10x faster!
```

**3. Liquid Clustering (Databricks Runtime 13.3+)**

```sql
-- Better than Z-order for high-cardinality keys
ALTER TABLE production.customers CLUSTER BY (customer_id);

-- Auto-optimizes on every write
```

**4. Small Source Optimization**

```sql
-- Broadcast small source table (< 10MB)
MERGE INTO production.customers AS target
USING (SELECT /*+ BROADCAST */ * FROM staging.cdc_events) AS source
ON target.customer_id = source.customer_id
...
```

---

#### **Common MERGE Pitfalls**

**❌ Pitfall 1: Duplicate Source Keys**

```sql
-- BAD: Source has duplicates
MERGE INTO target
USING source ON target.id = source.id  -- ❌ Error if source has duplicate IDs!
...

-- GOOD: Deduplicate source first
WITH deduped_source AS (
  SELECT id, name,
    ROW_NUMBER() OVER (PARTITION BY id ORDER BY timestamp DESC) AS rn
  FROM source
)
MERGE INTO target
USING deduped_source
ON target.id = deduped_source.id
  AND deduped_source.rn = 1
...
```

**❌ Pitfall 2: Not Handling NULLs in Join**

```sql
-- BAD: NULL keys don't match
ON target.customer_id = source.customer_id  -- ❌ NULLs won't match!

-- GOOD: Handle NULLs explicitly
ON target.customer_id = source.customer_id
  OR (target.customer_id IS NULL AND source.customer_id IS NULL)
```

**❌ Pitfall 3: Full Table Rewrites**

```sql
-- BAD: Updates all rows (full rewrite!)
UPDATE SET target.updated_at = CURRENT_TIMESTAMP()  -- ❌ Expensive!

-- GOOD: Only update changed rows
WHEN MATCHED AND (
  target.name != source.name OR
  target.email != source.email
) THEN UPDATE SET ...
```

---

#### **MERGE vs INSERT/UPDATE/DELETE**

**Without MERGE (Slow):**
```sql
-- Step 1: Handle inserts
INSERT INTO target SELECT * FROM source WHERE id NOT IN (SELECT id FROM target);

-- Step 2: Handle updates
UPDATE target SET ... FROM source WHERE target.id = source.id;

-- Step 3: Handle deletes
DELETE FROM target WHERE id IN (SELECT id FROM deletes);

-- ❌ 3 full table scans!
```

**With MERGE (Fast):**
```sql
MERGE INTO target USING source ON target.id = source.id
WHEN MATCHED THEN UPDATE SET ...
WHEN NOT MATCHED THEN INSERT ...;

-- ✅ Single table scan!
```

---

#### **Interview Answer Template**

1. **Explain MERGE benefits:** "Single operation handles INSERT/UPDATE/DELETE"
2. **Discuss deduplication:** "Always dedupe source on join keys"
3. **Mention optimization:** "Z-order on join keys, partition pruning"
4. **Real-world context:** "Used for CDC, SCD Type 1, upserts"
5. **Performance:** "3-10x faster than separate INSERT/UPDATE/DELETE"

````python
%sql
-- Demo: MERGE for CDC Upserts

-- Create target customer table
CREATE OR REPLACE TABLE workspace.default.customers_target (
  customer_id INT,
  name STRING,
  email STRING,
  status STRING,
  updated_at TIMESTAMP
);

-- Insert initial data
INSERT INTO workspace.default.customers_target VALUES
  (1, 'Alice', 'alice@email.com', 'active', CURRENT_TIMESTAMP()),
  (2, 'Bob', 'bob@email.com', 'active', CURRENT_TIMESTAMP()),
  (3, 'Carol', 'carol@email.com', 'inactive', CURRENT_TIMESTAMP());

-- Create CDC events (source)
CREATE OR REPLACE TABLE workspace.default.cdc_events (
  customer_id INT,
  name STRING,
  email STRING,
  status STRING,
  operation STRING,
  event_timestamp TIMESTAMP
);

INSERT INTO workspace.default.cdc_events VALUES
  (1, 'Alice Smith', 'alice@email.com', 'active', 'UPDATE', CURRENT_TIMESTAMP()),  -- Update
  (4, 'Dave', 'dave@email.com', 'active', 'INSERT', CURRENT_TIMESTAMP()),         -- Insert
  (3, 'Carol', 'carol@email.com', 'inactive', 'DELETE', CURRENT_TIMESTAMP());     -- Delete

-- MERGE: Apply CDC events
MERGE INTO workspace.default.customers_target AS target
USING workspace.default.cdc_events AS source
ON target.customer_id = source.customer_id

WHEN MATCHED AND source.operation = 'DELETE' THEN
  DELETE

WHEN MATCHED AND source.operation = 'UPDATE' THEN
  UPDATE SET
    target.name = source.name,
    target.email = source.email,
    target.status = source.status,
    target.updated_at = CURRENT_TIMESTAMP()

WHEN NOT MATCHED AND source.operation = 'INSERT' THEN
  INSERT (customer_id, name, email, status, updated_at)
  VALUES (source.customer_id, source.name, source.email, source.status, CURRENT_TIMESTAMP());

-- Verify results
SELECT * FROM workspace.default.customers_target ORDER BY customer_id;
/*
Expected:
1 | Alice Smith | alice@email.com | active     | <updated>
2 | Bob         | bob@email.com   | active     | <original>
4 | Dave        | dave@email.com  | active     | <new>
(Carol deleted)
*/

-- Check MERGE stats
DESCRIBE HISTORY workspace.default.customers_target LIMIT 1;
````

### ❓ Question 4: OPTIMIZE & VACUUM Maintenance

**Production Operations Interview:**
> "Your Delta table has degraded query performance over time. DESCRIBE DETAIL shows 50,000 small files. Explain the difference between OPTIMIZE and VACUUM, and when to use each. What are the risks?"

### ✅ Answer 4: OPTIMIZE & VACUUM

#### **The Small Files Problem**

**How it happens:**
- Streaming writes (micro-batches create small files)
- Many concurrent writers
- Frequent UPDATE/DELETE operations
- Partitioned tables with low cardinality

**Impact:**
- Slow query performance (metadata overhead)
- Excessive list operations on cloud storage
- Poor file pruning efficiency

---

#### **OPTIMIZE: Compaction & Layout**

**What it does:**
1. **File Compaction** - Combines small files into larger files (default 128MB)
2. **Z-Ordering** - Co-locates related data for better pruning
3. **Statistics Update** - Refreshes min/max stats for data skipping

**Syntax:**

```sql
-- Basic compaction
OPTIMIZE my_table;

-- Optimize specific partition
OPTIMIZE my_table WHERE date >= '2024-01-01';

-- Z-ordering (co-locate by frequently filtered columns)
OPTIMIZE my_table ZORDER BY (customer_id, date);

-- Liquid Clustering (auto-optimize, better than Z-order)
ALTER TABLE my_table CLUSTER BY (customer_id, date);
```

**When to OPTIMIZE:**
- After bulk INSERT/UPDATE/DELETE
- Streaming jobs (run daily/weekly)
- Query performance degradation
- Before critical reports/dashboards

**Cost/Benefit:**
- ✅ Faster queries (10-100x improvement possible)
- ❌ Storage temporarily increases (old + new files)
- ❌ Compute cost (rewriting data)
- ✅ Storage normalized after VACUUM

---

#### **VACUUM: Delete Old Files**

**What it does:**
- Permanently deletes data files not referenced by current version
- Removes files older than retention period
- Frees up storage space
- **Destroys time travel history!**

**Syntax:**

```sql
-- VACUUM with default retention (7 days)
VACUUM my_table;

-- VACUUM with custom retention (30 days)
VACUUM my_table RETAIN 720 HOURS;  -- 30 days

-- Dry run (see what would be deleted)
VACUUM my_table DRY RUN;
```

**When to VACUUM:**
- Weekly/monthly maintenance window
- Storage costs are high
- After OPTIMIZE (clean up old files)
- When time travel is no longer needed

**Risks:**

⚠️ **CRITICAL: VACUUM destroys time travel!**

```sql
-- Before VACUUM
SELECT * FROM my_table VERSION AS OF 100;  -- ✅ Works

-- Run VACUUM
VACUUM my_table RETAIN 168 HOURS;  -- Delete files > 7 days old

-- After VACUUM
SELECT * FROM my_table VERSION AS OF 100;  -- ❌ Error: files not found!
```

---

#### **OPTIMIZE vs VACUUM Decision Matrix**

| Goal | Command | Frequency | Impact |
|------|---------|-----------|--------|
| Fix slow queries | OPTIMIZE | After big writes | ✅ Faster reads |
| Reduce storage | VACUUM | Weekly/Monthly | ✅ Lower costs |
| Improve joins | OPTIMIZE ZORDER BY | Weekly | ✅ Better pruning |
| Enable time travel | Skip VACUUM | - | ✅ Keep history |
| Small files (< 1MB) | OPTIMIZE | Daily | ✅ Compaction |
| Free disk space | VACUUM | Monthly | ✅ Delete old files |

---

#### **Complete Maintenance Workflow**

```sql
-- Step 1: Check table health
DESCRIBE DETAIL my_table;
/*
Key metrics:
- numFiles: 50,000  ← Too many!
- sizeInBytes: 5GB
- minReaderVersion: 1
*/

-- Step 2: Check file size distribution
SELECT 
  CASE 
    WHEN size < 1024*1024 THEN '< 1MB'
    WHEN size < 10*1024*1024 THEN '1-10MB'
    WHEN size < 100*1024*1024 THEN '10-100MB'
    ELSE '> 100MB'
  END AS file_size_bucket,
  COUNT(*) AS file_count
FROM (
  SELECT input_file_name() AS file, COUNT(*) AS size
  FROM my_table
  GROUP BY file
)
GROUP BY file_size_bucket;

-- Step 3: OPTIMIZE (compaction + Z-order)
OPTIMIZE my_table ZORDER BY (customer_id, date);

-- Step 4: Verify improvement
DESCRIBE DETAIL my_table;
/*
After OPTIMIZE:
- numFiles: 40  ← Much better!
- sizeInBytes: 5GB (same data)
*/

-- Step 5: VACUUM (after retention period)
VACUUM my_table RETAIN 168 HOURS;  -- 7 days

-- Step 6: Monitor ongoing
SELECT 
  CAST(timestamp AS DATE) AS date,
  operation,
  operationMetrics.numFiles AS files_modified,
  operationMetrics.numOutputRows AS rows_modified
FROM (DESCRIBE HISTORY my_table)
WHERE operation IN ('OPTIMIZE', 'VACUUM')
ORDER BY timestamp DESC;
```

---

#### **Auto Optimize (Best Practice)**

```sql
-- Enable Auto Optimize on table creation
CREATE TABLE my_table (
  id INT,
  data STRING
)
USING DELTA
TBLPROPERTIES (
  'delta.autoOptimize.optimizeWrite' = 'true',  -- Optimize file size on write
  'delta.autoOptimize.autoCompact' = 'true'     -- Auto-compact small files
);

-- Enable on existing table
ALTER TABLE my_table SET TBLPROPERTIES (
  'delta.autoOptimize.optimizeWrite' = 'true',
  'delta.autoOptimize.autoCompact' = 'true'
);
```

**Auto Optimize Benefits:**
- Prevents small file problem
- No manual OPTIMIZE needed
- Small write overhead (~10%)
- Ideal for streaming tables

---

#### **Liquid Clustering vs Z-Ordering**

**Z-Ordering (Legacy):**
```sql
OPTIMIZE my_table ZORDER BY (col1, col2);
-- ❌ Manual maintenance
-- ❌ Order matters (col1 > col2 priority)
-- ❌ Full table rewrite
```

**Liquid Clustering (Modern):**
```sql
ALTER TABLE my_table CLUSTER BY (col1, col2);
-- ✅ Auto-optimizes on every write
-- ✅ Order doesn't matter
-- ✅ Incremental updates
-- ✅ Better for high-cardinality columns
```

**When to use:**
- **Z-Order:** Low-cardinality columns (< 1000 distinct values)
- **Liquid Clustering:** High-cardinality columns (IDs, timestamps)

**A:** "Yes, but with caution:
- Readers on current version: ✅ Safe
- Readers on old versions: ❌ May fail if files deleted

**A:** "VACUUM is atomic at the file level:

No data corruption risk, but storage may not be fully reclaimed until re-run."

### ❓ Question 5: Partitioning Strategy

**System Design Interview:**
> "You have a 5TB events table with 10B rows, queried primarily by date and customer_id. How would you partition this table? What are the trade-offs between partitioning by date, customer_id, or both?"

### ✅ Answer 5: Partitioning Strategy

#### **What is Partitioning?**

Physical organization of data into subdirectories:

```
/table/
  date=2024-01-01/
    part-00000.parquet
    part-00001.parquet
  date=2024-01-02/
    part-00000.parquet
```

**Benefits:**
- **Partition pruning** - Skip irrelevant partitions
- **Parallel processing** - Process partitions independently
- **Incremental updates** - Overwrite single partition

---

#### **Partitioning Decision Framework**

**Rule of Thumb:**
- **Target partition size:** 100MB - 1GB per partition
- **Max partitions:** < 10,000 per table
- **Cardinality:** Use low-cardinality columns (< 1000 distinct values)

---

#### **Option 1: Partition by Date**

```sql
CREATE TABLE events (
  event_id BIGINT,
  customer_id INT,
  event_type STRING,
  amount DECIMAL(10,2),
  event_date DATE
)
USING DELTA
PARTITIONED BY (event_date);
```

**Pros:**
- ✅ Queries with date filters are fast
- ✅ Easy incremental updates (daily batches)
- ✅ Simple retention (DROP old partitions)
- ✅ Predictable partition count (365 per year)

**Cons:**
- ❌ Queries by customer_id scan all partitions
- ❌ Partition size varies (weekends vs weekdays)
- ❌ Hot partitions (recent dates accessed more)

**Best for:**
- Time-series data
- Regular date-range queries
- Daily/hourly ETL patterns

---

#### **Option 2: Partition by customer_id**

```sql
PARTITIONED BY (customer_id)
```

**Pros:**
- ✅ Customer-specific queries are fast
- ✅ Parallel processing by customer

**Cons:**
- ❌ ⚠️ **Too many partitions!** (10M customers = 10M partitions)
- ❌ Small files (few events per customer)
- ❌ Metadata explosion (listing 10M directories)
- ❌ Date queries scan all partitions

**Verdict:** ❌ **Don't do this!** High-cardinality partitioning is an anti-pattern.

---

#### **Option 3: Multi-Level Partitioning (Recommended)**

```sql
-- Partition by date, bucket by customer_id
CREATE TABLE events (
  event_id BIGINT,
  customer_id INT,
  event_type STRING,
  amount DECIMAL(10,2)
)
USING DELTA
PARTITIONED BY (event_date)
CLUSTER BY (customer_id);  -- Liquid Clustering
```

**Or using Hive-style partitioning:**
```sql
PARTITIONED BY (year INT, month INT, day INT)
```

**Pros:**
- ✅ Date pruning (partition level)
- ✅ Customer pruning (data skipping via clustering)
- ✅ Balanced partition sizes
- ✅ Flexible querying

**Cons:**
- ❌ Slightly more complex
- ❌ Need to maintain clustering

---

#### **Option 4: No Partitioning + Liquid Clustering**

```sql
CREATE TABLE events (
  ...
)
USING DELTA
CLUSTER BY (event_date, customer_id);  -- No physical partitions
```

**When to use:**
- Table < 1TB
- Multiple query patterns
- High-cardinality keys
- Want automatic optimization

**Pros:**
- ✅ Automatic file organization
- ✅ No partition explosion
- ✅ Flexible querying
- ✅ Works for any column combination

**Cons:**
- ❌ Requires Databricks Runtime 13.3+
- ❌ Less explicit than partitioning

### ❓ Question 6: Broadcast vs Shuffle Joins

**Performance Tuning Interview:**
> "You're joining a 1TB fact table with a 10MB dimension table. The query takes 45 minutes. Explain the difference between broadcast joins and shuffle joins, and how you'd optimize this query."

### ✅ Answer 6: Broadcast vs Shuffle Joins

#### **Join Strategies in Spark SQL**

**1. Shuffle Join (Sort-Merge Join)**
- Both tables shuffled across network
- Data redistributed by join key
- Expensive for large tables

**2. Broadcast Join (Broadcast Hash Join)**
- Small table broadcasted to all nodes
- No shuffle of large table
- 10-100x faster for small tables

---

#### **Shuffle Join Explained**

```
Large Table (1TB)          Small Table (10MB)
[Node 1] [Node 2] [Node 3]  [Node 1] [Node 2]
     ↓         ↓         ↓        ↓         ↓
     Shuffle by join key (expensive!)
     ↓
[Node 1] [Node 2] [Node 3]
Join co-located data
```

**Cost:** O(N + M) network transfer (both tables)

---

#### **Broadcast Join Explained**

```
Small Table (10MB)         Large Table (1TB)
[Node 1]                   [Node 1] [Node 2] [Node 3]
     ↓                          ↓         ↓         ↓
  Broadcast to all nodes   No shuffle!
     ↓
[Node 1] [Node 2] [Node 3]
Join locally with broadcasted table
```

**Cost:** O(M) network transfer (small table only)

---

#### **When Spark Chooses Broadcast Join**

Automatic broadcast if:
```sql
-- Default threshold: 10MB
SET spark.sql.autoBroadcastJoinThreshold = 10485760;  -- 10MB

-- Query with small table
SELECT f.*, d.dimension_name
FROM fact_table f
JOIN dimension_table d ON f.dimension_id = d.dimension_id;
-- ✅ Auto-broadcast if dimension_table < 10MB
```

---

#### **Manual Broadcast Hint**

```sql
-- Force broadcast join
SELECT /*+ BROADCAST(d) */ f.*, d.dimension_name
FROM fact_table f
JOIN dimension_table d ON f.dimension_id = d.dimension_id;

-- Alternative syntax
SELECT f.*, d.dimension_name
FROM fact_table f
JOIN dimension_table d /*+ BROADCAST */ 
  ON f.dimension_id = d.dimension_id;
```

**Use when:**
- Table > 10MB but still broadcastable (< 1GB)
- Statistics are stale (Spark underestimates size)
- Testing performance

---

#### **Optimization Example: 1TB + 10MB**

**Before (Shuffle Join - 45 minutes):**
```sql
SELECT 
  f.order_id,
  f.amount,
  c.customer_name,
  c.customer_tier
FROM orders f  -- 1TB
JOIN customers c  -- 10MB
  ON f.customer_id = c.customer_id;

-- Execution plan:
-- SortMergeJoin [customer_id]
--   Exchange hashpartitioning(customer_id)  ← Shuffle 1TB!
--   Exchange hashpartitioning(customer_id)  ← Shuffle 10MB
```

**After (Broadcast Join - 2 minutes):**
```sql
SELECT /*+ BROADCAST(c) */
  f.order_id,
  f.amount,
  c.customer_name,
  c.customer_tier
FROM orders f  -- 1TB
JOIN customers c  -- 10MB
  ON f.customer_id = c.customer_id;

-- Execution plan:
-- BroadcastHashJoin [customer_id]
--   FileScan (orders)  ← No shuffle!
--   BroadcastExchange  ← 10MB broadcast
```

**Improvement:** 45 min → 2 min (22x faster!)

---

#### **Broadcast Join Limitations**

**1. Size Limit**
```sql
-- Default: 8GB driver memory
-- Broadcast table must fit in driver memory + executor memory

-- Increase broadcast threshold (risky!)
SET spark.sql.autoBroadcastJoinThreshold = 1073741824;  -- 1GB
```

⚠️ **Risk:** OutOfMemoryError if table too large

**2. Not Suitable for:**
- Large-large joins (both tables > 1GB)
- Skewed data (some keys have 1000x more rows)
- Full outer joins (Spark limitation)

#### **Advanced: Multiple Small Tables**


#### **When Broadcast Fails**

**Problem:** Table larger than driver memory

**Solutions:**

1. **Filter Before Join:**
```sql
WITH small_subset AS (
  SELECT * FROM dimension_table 
  WHERE is_active = TRUE  -- Reduce size

```

2. **Bucketing (Pre-partitioned Join):**
```sql

USING DELTA
CLUSTERED BY (customer_id) INTO 200 BUCKETS;

-- Join without shuffle (bucketed join)
SELECT * FROM fact_table f
JOIN customer_table c ON f.customer_id = c.customer_id;
```

3. **Increase Driver Memory:**
```python

spark.driver.memory = 32g
spark.sql.autoBroadcastJoinThreshold = 2147483648  # 2GB
```

#### **EXPLAIN Variants**

```sql
-- Logical plan only
EXPLAIN SELECT * FROM table WHERE x = 1;

EXPLAIN FORMATTED SELECT * FROM table WHERE x = 1;
```

---

#### **Key Plan Elements to Look For**

**🔴 RED FLAGS (Performance Killers):**

1. **Exchange (Shuffle Operations)**
   ```
   Exchange hashpartitioning(customer_id)
   ← Data shuffled across network (SLOW!)
   ```

2. **SortMergeJoin on Large Tables**
   ```
   SortMergeJoin [customer_id]
   ← Should use BroadcastHashJoin for small tables
   ```

3. **Cartesian Product**
   ```
   CartesianProduct
   ← N × M rows! Missing join condition?
   ```

4. **Full Table Scan (No Partition Pruning)**
   ```
   FileScan parquet ... PartitionFilters: []
   ← Scanning all partitions! Add WHERE clause?
   ```

**🟢 GOOD SIGNS (Optimized):**

1. **BroadcastHashJoin**
   ```
   BroadcastHashJoin [customer_id]
   ← Small table broadcasted (FAST!)
   ```

2. **PartitionFilters**
   ```
   PartitionFilters: [isnotnull(date), (date = 2024-01-15)]
   ← Only scanning relevant partitions!
   ```

3. **PushedFilters**
   ```
   PushedFilters: [IsNotNull(amount), GreaterThan(amount, 100)]
   ← Filters pushed to storage layer (data skipping)
   ```

4. **ColumnarToRow (Vectorized Execution)**
   ```
   ColumnarToRow
   ← Using Photon/Vectorized execution (FAST!)
   ```

#### **Real Example: Slow Query Diagnosis**

**Query:**
```sql
SELECT 
  o.order_id,
  c.customer_name,
  SUM(o.amount) AS total
FROM orders o
JOIN customers c ON o.customer_id = c.customer_id
WHERE o.order_date >= '2024-01-01'
GROUP BY o.order_id, c.customer_name;
```

**EXPLAIN Output (Slow):**
```

AdaptiveSparkPlan
+- HashAggregate
   +- Exchange hashpartitioning(order_id, customer_name)  ← 🔴 Shuffle!
      +- HashAggregate
         +- SortMergeJoin [customer_id]                    ← 🔴 Should be broadcast!
            +- Exchange hashpartitioning(customer_id)      ← 🔴 Shuffle orders!
            |  +- FileScan parquet orders
            |     PartitionFilters: [date >= 2024-01-01]   ← 🟢 Good!
            +- Exchange hashpartitioning(customer_id)      ← 🔴 Shuffle customers!

```

**Issues Found:**
1. ❌ SortMergeJoin (customers should be broadcast)
2. ❌ 3 Exchange operations (expensive shuffles)
3. ❌ Customers table shuffled (10MB table!)

**Optimized Query:**
```sql

```

**New EXPLAIN Output (Fast):**
```

HashAggregate
+- Exchange hashpartitioning(order_id, customer_name)  ← Still needed for GROUP BY
   +- HashAggregate
      +- BroadcastHashJoin [customer_id]                 ← 🟢 Broadcast join!

         |  PartitionFilters: [date >= 2024-01-01]
         +- BroadcastExchange                            ← 🟢 Only broadcast 10MB!
            +- FileScan parquet customers
```

**Result:** 10 min → 45 sec (13x faster!)

#### **Common Plan Patterns**

**Pattern 1: Skewed Join**
```

+- Exchange (shuffle write: 100GB, shuffle read: 100GB)
   +- Some partitions: 50GB (hot!)             ← 🔴 Skew!
   +- Most partitions: 100MB (cold)
```

**Fix:** Salt the join key
```sql
-- Add random salt to distribute hot keys
SELECT * FROM (
  SELECT *, FLOOR(RAND() * 10) AS salt FROM large_table
) JOIN (
  SELECT *, id AS salt FROM RANGE(10) CROSS JOIN small_table
) USING (customer_id, salt);
```

---

**Pattern 2: Missing Statistics**
```
Statistics(sizeInBytes=16.0 EiB)  ← 🔴 Default estimate (wrong!)
```

**Fix:** Collect stats
```sql
ANALYZE TABLE my_table COMPUTE STATISTICS;
ANALYZE TABLE my_table COMPUTE STATISTICS FOR ALL COLUMNS;
```

**Pattern 3: Unnecessary Repartition**
```
Repartition 200  ← 🔴 Manual repartition in code?
+- FileScan (already 200 partitions)
```

**Fix:** Remove explicit repartition
```python
# BAD
df.repartition(200).write.save(...)  # Already 200 partitions!

# GOOD
df.write.save(...)  # Let Spark decide
```

### ❓ Question 8: Unity Catalog Three-Level Namespace

**Architecture Interview:**
> "Explain Unity Catalog's three-level namespace (catalog.schema.table). How does it differ from Hive Metastore? What are the benefits for multi-tenant environments?"

### ✅ Answer 8: Unity Catalog Architecture

#### **Three-Level Namespace**

```
Unity Catalog Hierarchy:

Catalog (Database/Project)
  ├── Schema (Logical grouping)
  │   ├── Table
  │   ├── View
  │   ├── Function
  │   └── Volume (file storage)
  ├── Schema
  └── Schema

Example: production.sales.orders
         │         │      │
      catalog   schema  table
```

**Vs Hive Metastore (Two-Level):**
```
Hive Metastore:
  database.table  ← Only 2 levels
  │        │
schema    table
```

---

#### **Key Benefits**

**1. Multi-Tenancy**
```sql
-- Separate catalogs per team/environment
dev_team.analytics.sales       -- Dev catalog
staging_team.analytics.sales   -- Staging catalog
production.analytics.sales     -- Prod catalog

-- Isolate access by catalog
GRANT USE CATALOG dev_team TO `developers`;
GRANT USE CATALOG production TO `data_engineers`;
```

**2. Cross-Catalog Queries**
```sql
-- Join across catalogs
SELECT 
  prod.sales.orders.*,
  dev.experiments.features.*
FROM production.sales.orders prod
JOIN dev_team.experiments.features dev
  ON prod.customer_id = dev.customer_id;
```

**3. Centralized Governance**
- Single metastore across workspaces
- Unified lineage tracking
- Consistent access control
- Cross-cloud support (AWS, Azure, GCP)

**4. Fine-Grained Permissions**
```sql
-- Grant at any level
GRANT SELECT ON CATALOG production TO `analysts`;
GRANT SELECT ON SCHEMA production.sales TO `sales_team`;
GRANT SELECT ON TABLE production.sales.orders TO `order_analysts`;
```

#### **Unity Catalog vs Hive Metastore**

| Feature | Hive Metastore | Unity Catalog |
|---------|----------------|---------------|
| **Namespace** | 2-level (db.table) | 3-level (catalog.schema.table) |
| **Scope** | Per workspace | Cross-workspace |
| **Lineage** | None | Automatic column-level |
| **Governance** | Workspace-level | Account-level |
| **Delta Sharing** | Not supported | Built-in |
| **Volumes** | External locations only | Managed volumes |
| **Access Control** | Cluster-level | Fine-grained ACLs |
| **Audit Logging** | Limited | Comprehensive |
