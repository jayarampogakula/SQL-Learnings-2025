## 1️⃣ Introduction to SQL

### What is SQL?
**SQL (Structured Query Language)** is the standard language for interacting with **relational databases**.

### Key Concepts
* **Database**: A container that stores organized data
* **Table**: A structured collection of data organized in rows and columns
* **Row (Record)**: A single entry in a table
* **Column (Field)**: An attribute or property of the data
* **Schema**: The structure/blueprint of a database

### SQL Categories
| Category | Description | Commands |
|----------|-------------|----------|
| **DDL** (Data Definition Language) | Define database structure | `CREATE`, `ALTER`, `DROP`, `TRUNCATE` |
| **DML** (Data Manipulation Language) | Manipulate data | `INSERT`, `UPDATE`, `DELETE` |
| **DQL** (Data Query Language) | Query/retrieve data | `SELECT` |
| **DCL** (Data Control Language) | Control access | `GRANT`, `REVOKE` |
| **TCL** (Transaction Control Language) | Manage transactions | `COMMIT`, `ROLLBACK`, `SAVEPOINT` |

### Popular SQL Databases
* **Microsoft SQL Server**, **MySQL**, **PostgreSQL**, **Oracle Database**, **SQLite**, **Databricks SQL**

## 2️⃣ Database Fundamentals

### Data Types
Common SQL data types:

#### Numeric Types
* `INT` / `INTEGER` - Whole numbers (-2,147,483,648 to 2,147,483,647)
* `BIGINT` - Large integers
* `DECIMAL(p,s)` / `NUMERIC(p,s)` - Fixed precision (p = total digits, s = decimal places)
* `FLOAT` / `DOUBLE` - Approximate numeric values

#### String Types
* `VARCHAR(n)` - Variable-length string (up to n characters)
* `CHAR(n)` - Fixed-length string
* `TEXT` - Large text data

#### Date/Time Types
* `DATE` - Date only (YYYY-MM-DD)
* `TIME` - Time only (HH:MM:SS)
* `DATETIME` / `TIMESTAMP` - Date and time combined

#### Other Types
* `BOOLEAN` - True/False values
* `BLOB` - Binary data

### Keys
* **Primary Key**: Uniquely identifies each row (cannot be NULL, must be unique)
* **Foreign Key**: Creates relationship between tables by referencing another table's primary key

## 6️⃣ Filtering Data (WHERE Clause)

### Syntax
```sql
SELECT column1, column2
FROM table_name
WHERE condition;
```

### Comparison Operators
* `=` - Equal to
* `<>` or `!=` - Not equal to
* `>` - Greater than
* `<` - Less than
* `>=` - Greater than or equal to
* `<=` - Less than or equal to
* `BETWEEN` - Between a range
* `IN` - Matches any value in a list
* `LIKE` - Pattern matching with wildcards

### Logical Operators
* `AND` - All conditions must be true
* `OR` - At least one condition must be true
* `NOT` - Negates a condition

### Wildcard Characters (with LIKE)
* `%` - Matches any sequence of characters
* `_` - Matches any single character

````python
%sql
-- Simple comparison
SELECT * FROM Employees WHERE Department = 'IT';

-- Multiple conditions with AND
-- SELECT * FROM Employees WHERE Department = 'IT' AND Salary > 70000;

-- Multiple conditions with OR
-- SELECT * FROM Employees WHERE Department = 'IT' OR Department = 'HR';

-- Using IN operator (cleaner than multiple ORs)
-- SELECT * FROM Employees WHERE Department IN ('IT', 'HR', 'Finance');

-- Using BETWEEN for ranges
-- SELECT * FROM Employees WHERE Salary BETWEEN 65000 AND 75000;

-- Pattern matching with LIKE
-- SELECT * FROM Employees WHERE Email LIKE '%@company.com';
-- SELECT * FROM Employees WHERE FirstName LIKE 'J%';  -- Starts with J
````

## 9️⃣ Built-in Functions

### String Functions
* `UPPER(column)` - Convert to uppercase
* `LOWER(column)` - Convert to lowercase
* `LENGTH(column)` or `LEN(column)` - String length
* `CONCAT(str1, str2, ...)` - Concatenate strings
* `SUBSTRING(column, start, length)` - Extract substring
* `TRIM(column)` - Remove leading/trailing spaces
* `REPLACE(column, old, new)` - Replace text

### Numeric Functions
* `ROUND(number, decimals)` - Round to decimal places
* `FLOOR(number)` - Round down to integer
* `CEIL(number)` / `CEILING(number)` - Round up to integer
* `ABS(number)` - Absolute value
* `POWER(base, exponent)` - Raise to power

### Date/Time Functions
* `CURRENT_DATE()` or `GETDATE()` - Current date
* `NOW()` or `GETDATE()` - Current date and time
* `YEAR(date)`, `MONTH(date)`, `DAY(date)` - Extract parts
* `DATEDIFF(date1, date2)` - Difference between dates
* `DATEADD(interval, number, date)` - Add to date

### Aggregate Functions (used with GROUP BY)
* `COUNT(*)` - Count rows
* `SUM(column)` - Sum of values
* `AVG(column)` - Average value
* `MIN(column)` - Minimum value
* `MAX(column)` - Maximum value

````python
%sql
-- String functions
SELECT 
  FirstName,
  UPPER(FirstName) AS UpperName,
  LOWER(Email) AS LowerEmail,
  LENGTH(FirstName) AS NameLength,
  CONCAT(FirstName, ' ', LastName) AS FullName
FROM Employees;

-- Numeric functions
-- SELECT 
--   Salary,
--   ROUND(Salary, 0) AS RoundedSalary,
--   ROUND(Salary * 1.10, 2) AS SalaryWithRaise,
--   ABS(Salary - 70000) AS SalaryDiffFrom70K
-- FROM Employees;

-- Date functions
-- SELECT 
--   FirstName,
--   HireDate,
--   YEAR(HireDate) AS HireYear,
--   DATEDIFF(CURRENT_DATE(), HireDate) AS DaysSinceHire
-- FROM Employees;

-- Aggregate functions
-- SELECT 
--   COUNT(*) AS TotalEmployees,
--   AVG(Salary) AS AvgSalary,
--   MIN(Salary) AS MinSalary,
--   MAX(Salary) AS MaxSalary,
--   SUM(Salary) AS TotalPayroll
-- FROM Employees;
````

## 1️⃣1️⃣ Handling NULL Values

### What is NULL?
* `NULL` represents missing or unknown data
* `NULL` is NOT the same as 0, empty string '', or FALSE
* Cannot use `=` or `<>` with NULL

### Checking for NULL
```sql
WHERE column IS NULL
WHERE column IS NOT NULL
```

### Handling NULL Values
* `COALESCE(column, default)` - Returns first non-NULL value
* `IFNULL(column, default)` - MySQL version
* `ISNULL(column, default)` - SQL Server version
* `NVL(column, default)` - Oracle version

````python
%sql
-- Find employees without a manager
SELECT FirstName, LastName, ManagerID 
FROM Employees 
WHERE ManagerID IS NULL;

-- Find employees with a manager
-- SELECT FirstName, LastName, ManagerID 
-- FROM Employees 
-- WHERE ManagerID IS NOT NULL;

-- Replace NULL with default value
-- SELECT 
--   FirstName,
--   ManagerID,
--   COALESCE(ManagerID, 0) AS ManagerIDOrZero,
--   COALESCE(CAST(ManagerID AS STRING), 'No Manager') AS ManagerStatus
-- FROM Employees;
````

## 1️⃣2️⃣ Modifying Data (UPDATE)

### Syntax
```sql
UPDATE table_name
SET column1 = value1, column2 = value2, ...
```

### ⚠️ Important Warning
**ALWAYS use a WHERE clause with UPDATE!**
Without WHERE, ALL rows will be updated!

### Best Practices

## 1️⃣3️⃣ Deleting Data (DELETE)

### Syntax
```sql
DELETE FROM table_name
WHERE condition;
```

### ⚠️ Critical Warning
**ALWAYS use a WHERE clause with DELETE!**
Without WHERE, ALL rows will be deleted!

### DELETE vs TRUNCATE vs DROP
* `DELETE` - Removes specific rows, can be filtered, can be rolled back
* `TRUNCATE` - Removes all rows, faster than DELETE, cannot be rolled back
* `DROP` - Removes entire table structure and data

````python
%sql
-- Delete specific employee
DELETE FROM Employees 
WHERE EmployeeID = 6;

-- Delete with multiple conditions
-- DELETE FROM Employees 
-- WHERE Department = 'Marketing' AND Salary < 50000;

-- Delete based on subquery (advanced - will learn in Module 2)
-- DELETE FROM Employees 
-- WHERE EmployeeID IN (
--   SELECT EmployeeID FROM Employees WHERE HireDate < '2020-01-01'
-- );
````

## 1️⃣4️⃣ Altering Tables (ALTER)
