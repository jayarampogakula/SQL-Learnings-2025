## 📌 Section 1: SELECT & Filtering (4 Questions)

The SELECT statement is the foundation of SQL. Mastering filtering with WHERE is essential for efficient data retrieval.

### ❓ Question 1: Basic SELECT and Column Filtering

**Fundamental Interview Question:**
> "You have an employees table with columns: employee_id, first_name, last_name, email, salary, department, hire_date. Write a query to retrieve employee names (full name as one column), email, and salary for all employees earning more than $50,000, ordered by salary descending."

### ✅ Answer 1: SELECT Best Practices

#### **Solution:**

```sql
SELECT 
  CONCAT(first_name, ' ', last_name) AS full_name,
  email,
  salary
FROM employees
WHERE salary > 50000
ORDER BY salary DESC;
```

#### **Key Concepts:**

**1. Column Selection**
- ✅ **Good:** Select only needed columns (`first_name, email, salary`)
- ❌ **Bad:** Use `SELECT *` (slow, fragile to schema changes)

**2. Column Aliases**
```sql
-- Use AS for readability
CONCAT(first_name, ' ', last_name) AS full_name

-- AS is optional but recommended
CONCAT(first_name, ' ', last_name) full_name  -- Also valid
```

**3. WHERE Clause**
- Filters rows BEFORE aggregation
- Use comparison operators: `=, !=, <, >, <=, >=`
- Can combine with `AND, OR, NOT`

**4. ORDER BY**
```sql
-- Ascending (default)
ORDER BY salary
ORDER BY salary ASC

-- Descending
ORDER BY salary DESC

-- Multiple columns
ORDER BY department ASC, salary DESC
```

#### **Query Execution Order (CRITICAL!):**

```
1. FROM      → Identify source table
2. WHERE     → Filter rows
3. SELECT    → Choose columns (and compute expressions)
4. ORDER BY  → Sort results
```

**Why this matters:**
```sql
-- ❌ This FAILS (column alias not available in WHERE)
SELECT 
  salary * 1.1 AS new_salary
FROM employees
WHERE new_salary > 50000;  -- Error: new_salary doesn't exist yet!

-- ✅ This WORKS
SELECT 
  salary * 1.1 AS new_salary
FROM employees
WHERE salary * 1.1 > 50000;  -- Use expression directly

-- ✅ Or use subquery/CTE
WITH calc AS (
  SELECT salary * 1.1 AS new_salary
  FROM employees
)
SELECT * FROM calc
WHERE new_salary > 50000;
```

#### **Common Mistakes:**

**❌ Mistake 1: Using SELECT * **
```sql
-- Bad: Returns all columns (slow, wastes bandwidth)
SELECT * FROM employees WHERE salary > 50000;

-- Good: Select only what you need
SELECT first_name, salary FROM employees WHERE salary > 50000;
```

**❌ Mistake 2: Not using ORDER BY**
```sql
-- Bad: Order is non-deterministic!
SELECT * FROM employees LIMIT 10;  -- Which 10?

-- Good: Always ORDER BY when using LIMIT
SELECT * FROM employees ORDER BY hire_date DESC LIMIT 10;
```

**❌ Mistake 3: String comparison without proper quoting**
```sql
-- Bad: Syntax error
WHERE department = Engineering  -- Missing quotes!

-- Good
WHERE department = 'Engineering'
```

#### **Interview Follow-Up Questions:**

**Q: "What's the difference between WHERE and HAVING?"**

**A:** 
- **WHERE:** Filters rows BEFORE aggregation
- **HAVING:** Filters groups AFTER aggregation

```sql
-- WHERE: Filter individual rows
SELECT * FROM employees WHERE salary > 50000;

-- HAVING: Filter aggregated groups
SELECT department, AVG(salary)
FROM employees
GROUP BY department
HAVING AVG(salary) > 50000;
```

**Q: "Why avoid SELECT * in production?"**

**A:**
1. **Performance:** Returns unnecessary columns (network + memory overhead)
2. **Fragility:** Schema changes break dependent code
3. **Readability:** Unclear what data is actually needed
4. **Security:** May expose sensitive columns unintentionally

````python
%sql
-- Demo: Basic SELECT and Filtering

-- Create sample employees table
CREATE OR REPLACE TABLE workspace.default.employees (
  employee_id INT,
  first_name STRING,
  last_name STRING,
  email STRING,
  salary DECIMAL(10,2),
  department STRING,
  hire_date DATE
);

INSERT INTO workspace.default.employees VALUES
  (1, 'Alice', 'Johnson', 'alice@company.com', 75000, 'Engineering', '2022-01-15'),
  (2, 'Bob', 'Smith', 'bob@company.com', 55000, 'Sales', '2021-03-20'),
  (3, 'Carol', 'Davis', 'carol@company.com', 92000, 'Engineering', '2020-06-10'),
  (4, 'Dave', 'Wilson', 'dave@company.com', 48000, 'Marketing', '2023-02-01'),
  (5, 'Eve', 'Brown', 'eve@company.com', 68000, 'Sales', '2021-11-05');

-- Query: Employees earning > $50K, ordered by salary
SELECT 
  CONCAT(first_name, ' ', last_name) AS full_name,
  email,
  salary,
  department
FROM workspace.default.employees
WHERE salary > 50000
ORDER BY salary DESC;
````

### ❓ Question 2: Advanced WHERE Clause Filtering

**Interview Question:**
> "Write a query to find employees who:
> 1. Work in Engineering OR Sales department
> 2. Were hired in 2021 or 2022
> 3. Have salary between $50K and $80K
> 4. Email domain is 'company.com'"

### ✅ Answer 2: WHERE Clause Operators

#### **Solution:**

```sql
SELECT 
  first_name,
  last_name,
  department,
  salary,
  hire_date,
  email
FROM employees
WHERE 
  department IN ('Engineering', 'Sales')  -- Condition 1
  AND YEAR(hire_date) IN (2021, 2022)     -- Condition 2
  AND salary BETWEEN 50000 AND 80000      -- Condition 3
  AND email LIKE '%@company.com';         -- Condition 4
```

#### **WHERE Clause Operators:**

**1. IN Operator**
```sql
-- Check if value matches any in a list
WHERE department IN ('Engineering', 'Sales', 'Marketing')

-- Equivalent to:
WHERE department = 'Engineering' 
   OR department = 'Sales' 
   OR department = 'Marketing'
```

**2. BETWEEN Operator**
```sql
-- Inclusive range check
WHERE salary BETWEEN 50000 AND 80000

-- Equivalent to:
WHERE salary >= 50000 AND salary <= 80000

-- Also works with dates
WHERE hire_date BETWEEN '2021-01-01' AND '2022-12-31'
```

**3. LIKE Operator (Pattern Matching)**
```sql
-- % = any characters (0 or more)
WHERE email LIKE '%@company.com'     -- Ends with @company.com
WHERE name LIKE 'John%'              -- Starts with John
WHERE name LIKE '%Smith%'            -- Contains Smith

-- _ = exactly one character
WHERE code LIKE 'A_C'                -- ABC, ADC, AZC match

-- Case sensitivity (depends on database)
WHERE LOWER(email) LIKE '%@company.com'  -- Case-insensitive
```

**4. NOT Operator**
```sql
-- Negate conditions
WHERE department NOT IN ('HR', 'Legal')
WHERE salary NOT BETWEEN 40000 AND 60000
WHERE email NOT LIKE '%temp%'
```

**5. IS NULL / IS NOT NULL**
```sql
-- ❗ CRITICAL: NULL comparisons
WHERE email IS NULL       -- ✅ Correct
WHERE email = NULL        -- ❌ WRONG! Always returns FALSE

WHERE email IS NOT NULL   -- ✅ Correct
WHERE email != NULL       -- ❌ WRONG!
```

#### **Logical Operators:**

**AND (All conditions must be TRUE)**
```sql
WHERE salary > 50000 AND department = 'Engineering'
```

**OR (At least one condition must be TRUE)**
```sql
WHERE department = 'Engineering' OR department = 'Sales'
```

**Precedence (AND before OR)**
```sql
-- Ambiguous:
WHERE salary > 50000 OR department = 'Sales' AND is_manager = TRUE
-- Interpreted as: (salary > 50000) OR (department = 'Sales' AND is_manager = TRUE)

-- Explicit with parentheses:
WHERE (salary > 50000 OR department = 'Sales') AND is_manager = TRUE
```

#### **Common Mistakes:**

**❌ Mistake 1: Using = instead of IN**
```sql
-- Bad: Only checks first value
WHERE department = ('Engineering', 'Sales')  -- Syntax error!

-- Good
WHERE department IN ('Engineering', 'Sales')
```

**❌ Mistake 2: Incorrect NULL handling**
```sql
-- Bad: Never returns TRUE
WHERE email = NULL
WHERE email != NULL

-- Good
WHERE email IS NULL
WHERE email IS NOT NULL
```

**❌ Mistake 3: LIKE without wildcards**
```sql
-- Bad: Exact match, use = instead
WHERE email LIKE 'alice@company.com'

-- Good: Use = for exact match
WHERE email = 'alice@company.com'

-- Or use LIKE with pattern
WHERE email LIKE '%alice%'
```

#### **Performance Tips:**

1. **Put most selective filters first** (in some databases)
   ```sql
   WHERE employee_id = 123        -- High selectivity (1 row)
     AND department = 'Engineering' -- Lower selectivity (100s of rows)
   ```

2. **Avoid functions on indexed columns**
   ```sql
   -- Bad: Can't use index on hire_date
   WHERE YEAR(hire_date) = 2022
   
   -- Good: Can use index
   WHERE hire_date BETWEEN '2022-01-01' AND '2022-12-31'
   ```

3. **Use IN instead of multiple ORs**
   ```sql
   -- Less efficient
   WHERE dept = 'A' OR dept = 'B' OR dept = 'C'
   
   -- More efficient
   WHERE dept IN ('A', 'B', 'C')
   ```

````python
%sql
-- Demo: WHERE Clause Patterns

-- Filter with multiple conditions
SELECT 
  CONCAT(first_name, ' ', last_name) AS full_name,
  department,
  salary,
  hire_date
FROM workspace.default.employees
WHERE 
  department IN ('Engineering', 'Sales')
  AND YEAR(hire_date) BETWEEN 2021 AND 2022
  AND salary BETWEEN 50000 AND 80000
  AND email LIKE '%@company.com';

-- Pattern matching examples
SELECT first_name, email
FROM workspace.default.employees
WHERE email LIKE '%company.com';  -- Ends with company.com

-- NULL handling
SELECT first_name, department
FROM workspace.default.employees
WHERE department IS NOT NULL;
````

## 📌 Section 2: Aggregations & GROUP BY (4 Questions)

Aggregation functions (COUNT, SUM, AVG, MIN, MAX) and GROUP BY are essential for data analysis and reporting.

### ❓ Question 3: Aggregate Functions

**Interview Question:**
> "Write queries to find:
> 1. Total number of employees
> 2. Average salary across all employees
> 3. Highest and lowest salaries
> 4. Total salary expense per department"

### ✅ Answer 3: Aggregate Functions

#### **Solutions:**

```sql
-- 1. Total number of employees
SELECT COUNT(*) AS total_employees
FROM employees;

-- 2. Average salary
SELECT AVG(salary) AS average_salary
FROM employees;

-- 3. Highest and lowest salaries
SELECT 
  MAX(salary) AS highest_salary,
  MIN(salary) AS lowest_salary
FROM employees;

-- 4. Total salary expense per department
SELECT 
  department,
  SUM(salary) AS total_salary_expense,
  COUNT(*) AS employee_count
FROM employees
GROUP BY department;
```

#### **Aggregate Functions:**

**COUNT()**
```sql
-- Count all rows (including NULLs)
SELECT COUNT(*) FROM employees;  -- Returns 5

-- Count non-NULL values in a column
SELECT COUNT(email) FROM employees;  -- Returns 4 if 1 email is NULL

-- Count distinct values
SELECT COUNT(DISTINCT department) FROM employees;  -- Returns 3
```

**SUM()**
```sql
-- Total of all salaries
SELECT SUM(salary) FROM employees;

-- ⚠️ NULL values are ignored
SELECT SUM(bonus) FROM employees;  -- NULL bonuses not included
```

**AVG()**
```sql
-- Average salary
SELECT AVG(salary) FROM employees;

-- ⚠️ Divides by count of non-NULL values
SELECT AVG(bonus) FROM employees;  -- Only averages non-NULL bonuses
```

**MIN() / MAX()**
```sql
-- Works on numbers, dates, strings
SELECT MIN(salary), MAX(salary) FROM employees;
SELECT MIN(hire_date), MAX(hire_date) FROM employees;
SELECT MIN(last_name), MAX(last_name) FROM employees;  -- Alphabetical
```

#### **GROUP BY (Critical Concept!):**

**Basic Grouping:**
```sql
SELECT 
  department,
  COUNT(*) AS employee_count,
  AVG(salary) AS avg_salary
FROM employees
GROUP BY department;
```

**How GROUP BY Works:**
1. Partition rows into groups based on GROUP BY columns
2. Apply aggregate function to each group
3. Return one row per group

**Multi-Column Grouping:**
```sql
SELECT 
  department,
  YEAR(hire_date) AS hire_year,
  COUNT(*) AS new_hires
FROM employees
GROUP BY department, YEAR(hire_date);
```

#### **GROUP BY Rules:**

**Rule 1: SELECT columns must be in GROUP BY or aggregated**
```sql
-- ❌ Bad: first_name not in GROUP BY or aggregate
SELECT department, first_name, COUNT(*)
FROM employees
GROUP BY department;  -- Error!

-- ✅ Good: Only grouped or aggregated columns
SELECT department, COUNT(*)
FROM employees
GROUP BY department;
```

**Rule 2: WHERE filters before GROUP BY**
```sql
SELECT department, AVG(salary)
FROM employees
WHERE salary > 50000  -- Filter rows first
GROUP BY department;  -- Then group
```

**Rule 3: HAVING filters after GROUP BY**
```sql
SELECT department, AVG(salary) AS avg_sal
FROM employees
GROUP BY department
HAVING AVG(salary) > 60000;  -- Filter groups
```

#### **WHERE vs HAVING:**

| Aspect | WHERE | HAVING |
|--------|-------|--------|
| **When** | Before GROUP BY | After GROUP BY |
| **Filters** | Individual rows | Aggregated groups |
| **Can use aggregates** | No | Yes |
| **Performance** | Faster (reduces rows early) | Slower (filters after aggregation) |

**Example:**
```sql
SELECT 
  department,
  COUNT(*) AS employee_count,
  AVG(salary) AS avg_salary
FROM employees
WHERE hire_date >= '2021-01-01'  -- WHERE: Filter rows
GROUP BY department
HAVING COUNT(*) >= 2;            -- HAVING: Filter groups
```

#### **Common Mistakes:**

**❌ Mistake 1: Mixing aggregates and non-aggregates**
```sql
-- Bad: Can't mix without GROUP BY
SELECT department, COUNT(*) FROM employees;  -- Error!

-- Good: Group by non-aggregate columns
SELECT department, COUNT(*) FROM employees GROUP BY department;
```

**❌ Mistake 2: Using WHERE instead of HAVING**
```sql
-- Bad: Can't use aggregate in WHERE
SELECT department, AVG(salary)
FROM employees
WHERE AVG(salary) > 60000  -- Error!
GROUP BY department;

-- Good: Use HAVING for aggregate filters
SELECT department, AVG(salary)
FROM employees
GROUP BY department
HAVING AVG(salary) > 60000;
```

**❌ Mistake 3: Forgetting COUNT(*) vs COUNT(column)**
```sql
-- COUNT(*) includes rows with NULL
SELECT COUNT(*) FROM employees;  -- 5 rows

-- COUNT(column) excludes NULL values
SELECT COUNT(email) FROM employees;  -- 4 rows (1 NULL email)
```

#### **Query Execution Order with GROUP BY:**

```
1. FROM      → Get source table
2. WHERE     → Filter rows
3. GROUP BY  → Group rows
4. HAVING    → Filter groups
5. SELECT    → Select columns/aggregates
6. ORDER BY  → Sort results
7. LIMIT     → Limit results
```

````python
%sql
-- Demo: Aggregate Functions and GROUP BY

-- Basic aggregations
SELECT 
  COUNT(*) AS total_employees,
  AVG(salary) AS average_salary,
  MAX(salary) AS highest_salary,
  MIN(salary) AS lowest_salary,
  SUM(salary) AS total_payroll
FROM workspace.default.employees;

-- GROUP BY department
SELECT 
  department,
  COUNT(*) AS employee_count,
  AVG(salary) AS avg_salary,
  MAX(salary) AS max_salary,
  MIN(salary) AS min_salary
FROM workspace.default.employees
GROUP BY department
ORDER BY avg_salary DESC;

-- WHERE + GROUP BY + HAVING
SELECT 
  department,
  COUNT(*) AS employee_count,
  AVG(salary) AS avg_salary
FROM workspace.default.employees
WHERE salary > 50000  -- Filter rows before grouping
GROUP BY department
HAVING COUNT(*) >= 2  -- Filter groups after aggregation
ORDER BY avg_salary DESC;
````

## 📌 Section 3: JOIN Operations (5 Questions)

JOINs combine data from multiple tables. Understanding JOIN types is crucial for working with relational databases.

### ❓ Question 4: INNER JOIN

**Fundamental Interview Question:**
> "You have two tables:
> - employees: employee_id, first_name, last_name, department_id
> - departments: department_id, department_name, location
>
> Write a query to show each employee's name with their department name and location. Only include employees who have a department assigned."

### ✅ Answer 4: INNER JOIN

#### **Solution:**

```sql
SELECT 
  e.employee_id,
  e.first_name,
  e.last_name,
  d.department_name,
  d.location
FROM employees e
INNER JOIN departments d 
  ON e.department_id = d.department_id;
```

#### **INNER JOIN Explanation:**

**What INNER JOIN Does:**
- Returns rows where there's a match in BOTH tables
- Excludes rows without a match
- Most common JOIN type

**Visual Example:**
```
employees               departments
┌──────────────┐  ┌──────────────────┐
│ emp_id | dept_id │  │ dept_id | dept_name │
├──────────────┤  ├──────────────────┤
│ 1      | 10      │  │ 10      | Sales     │
│ 2      | 20      │  │ 20      | IT        │
│ 3      | NULL    │  │ 30      | HR        │
└──────────────┘  └──────────────────┘

INNER JOIN Result:
┌──────────────────────────────┐
│ emp_id | dept_id | dept_name │
├──────────────────────────────┤
│ 1      | 10      | Sales     │  ✅ Match
│ 2      | 20      | IT        │  ✅ Match
└──────────────────────────────┘

Excluded:
- emp_id 3 (dept_id NULL) ❌ No match
- dept_id 30 (HR) ❌ No employees
```

#### **JOIN Syntax Variants:**

**1. Explicit JOIN (Recommended)**
```sql
SELECT e.*, d.department_name
FROM employees e
INNER JOIN departments d ON e.department_id = d.department_id;
```

**2. Implicit JOIN (Old style, avoid)**
```sql
-- Comma-separated tables with WHERE condition
SELECT e.*, d.department_name
FROM employees e, departments d
WHERE e.department_id = d.department_id;
```

**3. USING Clause (when column names match)**
```sql
SELECT e.*, d.department_name
FROM employees e
INNER JOIN departments d USING (department_id);
```

**4. NATURAL JOIN (auto-joins on same column names, risky!)**
```sql
-- Joins on ALL matching column names
SELECT *
FROM employees
NATURAL JOIN departments;  -- ⚠️ Avoid! Implicit behavior
```

#### **Table Aliases (Best Practice):**

```sql
-- Use short aliases for readability
SELECT 
  e.first_name,     -- Clear: from employees
  d.department_name -- Clear: from departments
FROM employees e    -- Alias 'e'
INNER JOIN departments d  -- Alias 'd'
  ON e.department_id = d.department_id;
```

#### **Multiple JOINs:**

```sql
SELECT 
  e.first_name,
  d.department_name,
  l.city,
  l.country
FROM employees e
INNER JOIN departments d ON e.department_id = d.department_id
INNER JOIN locations l ON d.location_id = l.location_id;
```

#### **Self JOIN (Joining Table to Itself):**

```sql
-- Find employees and their managers
SELECT 
  emp.first_name AS employee,
  mgr.first_name AS manager
FROM employees emp
INNER JOIN employees mgr 
  ON emp.manager_id = mgr.employee_id;
```

#### **Common Mistakes:**

**❌ Mistake 1: Missing ON condition (Cartesian product!)**
```sql
-- Bad: Creates N × M rows!
SELECT * FROM employees e, departments d;  -- 1000 employees × 50 depts = 50,000 rows!

-- Good: Use JOIN with ON
SELECT * FROM employees e
INNER JOIN departments d ON e.department_id = d.department_id;
```

**❌ Mistake 2: Ambiguous column names**
```sql
-- Bad: Which table's employee_id?
SELECT employee_id, department_name
FROM employees e
INNER JOIN departments d ON e.department_id = d.department_id;  -- Error if both tables have employee_id!

-- Good: Always qualify columns
SELECT e.employee_id, d.department_name
FROM employees e
INNER JOIN departments d ON e.department_id = d.department_id;
```

**❌ Mistake 3: Using INNER JOIN when LEFT JOIN is needed**
```sql
-- Bad: Excludes employees without departments
SELECT e.*, d.department_name
FROM employees e
INNER JOIN departments d ON e.department_id = d.department_id;

-- Good: Include all employees
SELECT e.*, d.department_name
FROM employees e
LEFT JOIN departments d ON e.department_id = d.department_id;
```

````python
%sql
-- Demo: INNER JOIN

-- Create departments table
CREATE OR REPLACE TABLE workspace.default.departments (
  department_id INT,
  department_name STRING,
  location STRING
);

INSERT INTO workspace.default.departments VALUES
  (1, 'Engineering', 'San Francisco'),
  (2, 'Sales', 'New York'),
  (3, 'Marketing', 'Austin'),
  (4, 'HR', 'Chicago');  -- No employees in HR

-- Update employees to use department_id
DROP TABLE IF EXISTS workspace.default.employees;

CREATE OR REPLACE TABLE workspace.default.employees (
  employee_id INT,
  first_name STRING,
  last_name STRING,
  email STRING,
  salary DECIMAL(10,2),
  department_id INT,
  hire_date DATE
);

INSERT INTO workspace.default.employees VALUES
  (1, 'Alice', 'Johnson', 'alice@company.com', 75000, 1, '2022-01-15'),
  (2, 'Bob', 'Smith', 'bob@company.com', 55000, 2, '2021-03-20'),
  (3, 'Carol', 'Davis', 'carol@company.com', 92000, 1, '2020-06-10'),
  (4, 'Dave', 'Wilson', 'dave@company.com', 48000, 3, '2023-02-01'),
  (5, 'Eve', 'Brown', 'eve@company.com', 68000, 2, '2021-11-05'),
  (6, 'Frank', 'Miller', 'frank@company.com', 82000, NULL, '2022-08-15');  -- No department

-- INNER JOIN: Only employees with departments
SELECT 
  e.employee_id,
  CONCAT(e.first_name, ' ', e.last_name) AS full_name,
  e.salary,
  d.department_name,
  d.location
FROM workspace.default.employees e
INNER JOIN workspace.default.departments d 
  ON e.department_id = d.department_id
ORDER BY d.department_name, e.last_name;
````

### ❓ Question 5: LEFT JOIN (LEFT OUTER JOIN)

**Interview Question:**
> "Modify the previous query to include ALL employees, even those without a department assigned. Show 'No Department' for employees with NULL department_id."

### ✅ Answer 5: LEFT JOIN

#### **Solution:**

```sql
SELECT 
  e.employee_id,
  e.first_name,
  e.last_name,
  COALESCE(d.department_name, 'No Department') AS department,
  COALESCE(d.location, 'N/A') AS location
FROM employees e
LEFT JOIN departments d 
  ON e.department_id = d.department_id;
```

#### **LEFT JOIN Explanation:**

**What LEFT JOIN Does:**
- Returns ALL rows from LEFT table (employees)
- Matching rows from RIGHT table (departments)
- NULLs for right table columns when no match

**Visual Example:**
```
employees (LEFT)        departments (RIGHT)
┌──────────────┐  ┌──────────────────┐
│ emp_id | dept_id │  │ dept_id | dept_name │
├──────────────┤  ├──────────────────┤
│ 1      | 10      │  │ 10      | Sales     │
│ 2      | 20      │  │ 20      | IT        │
│ 3      | NULL    │  │ 30      | HR        │
└──────────────┘  └──────────────────┘

LEFT JOIN Result:
┌──────────────────────────────┐
│ emp_id | dept_id | dept_name │
├──────────────────────────────┤
│ 1      | 10      | Sales     │  ✅ Match
│ 2      | 20      | IT        │  ✅ Match
│ 3      | NULL    | NULL      │  ⚠️ No match
└──────────────────────────────┘

All employees included!
HR department (30) still excluded (no match in left table)
```

#### **When to Use LEFT JOIN:**

1. **Find unmatched records**
   ```sql
   -- Employees without departments
   SELECT e.*
   FROM employees e
   LEFT JOIN departments d ON e.department_id = d.department_id
   WHERE d.department_id IS NULL;
   ```

2. **Include all primary records**
   ```sql
   -- All customers, with or without orders
   SELECT c.*, o.order_count
   FROM customers c
   LEFT JOIN (
     SELECT customer_id, COUNT(*) AS order_count
     FROM orders
     GROUP BY customer_id
   ) o ON c.customer_id = o.customer_id;
   ```

3. **Optional relationships**
   ```sql
   -- All employees, with manager name if they have one
   SELECT e.name, m.name AS manager_name
   FROM employees e
   LEFT JOIN employees m ON e.manager_id = m.employee_id;
   ```

#### **COALESCE for NULL Handling:**

```sql
-- Replace NULL with default value
COALESCE(d.department_name, 'No Department')
COALESCE(d.location, 'Unknown')

-- Returns first non-NULL value
COALESCE(col1, col2, col3, 'Default')
```

#### **Other JOIN Types:**

**RIGHT JOIN (Same as LEFT JOIN, flipped)**
```sql
-- These are equivalent:
SELECT * FROM employees e LEFT JOIN departments d ON e.dept_id = d.dept_id;
SELECT * FROM departments d RIGHT JOIN employees e ON e.dept_id = d.dept_id;
```

**FULL OUTER JOIN (All rows from both tables)**
```sql
-- Include all employees AND all departments
SELECT 
  COALESCE(e.employee_id, 'N/A') AS emp_id,
  COALESCE(e.first_name, 'No Employee') AS name,
  COALESCE(d.department_name, 'No Department') AS dept
FROM employees e
FULL OUTER JOIN departments d ON e.department_id = d.department_id;
```

**CROSS JOIN (Cartesian product)**
```sql
-- Every employee paired with every department (N × M rows)
SELECT e.name, d.department_name
FROM employees e
CROSS JOIN departments d;

-- Use case: Generate all date-product combinations
SELECT d.date, p.product_id
FROM date_dimension d
CROSS JOIN products p;
```

#### **JOIN Type Decision Tree:**

```
Need all rows from both tables?
  YES → FULL OUTER JOIN
  NO  ↓

Need all rows from left table?
  YES → LEFT JOIN
  NO  ↓

Need all rows from right table?
  YES → RIGHT JOIN (or swap and use LEFT)
  NO  ↓

Only matching rows?
  YES → INNER JOIN

Need every combination?
  YES → CROSS JOIN
```

#### **Performance Considerations:**

**LEFT JOIN vs INNER JOIN:**
- INNER JOIN is usually faster (smaller result set)
- Use LEFT JOIN only when you need unmatched rows

**Filtering with LEFT JOIN:**
```sql
-- ❌ Bad: Turns LEFT JOIN into INNER JOIN
SELECT *
FROM employees e
LEFT JOIN departments d ON e.department_id = d.department_id
WHERE d.department_name = 'Sales';  -- Excludes NULLs!

-- ✅ Good: Filter in ON clause to keep LEFT JOIN behavior
SELECT *
FROM employees e
LEFT JOIN departments d 
  ON e.department_id = d.department_id 
  AND d.department_name = 'Sales';  -- Keep all employees
```

Subqueries and Common Table Expressions (CTEs) help break complex queries into logical steps.

### ❓ Question 6: Subqueries

**Interview Question:**
> "Find all employees whose salary is above the average salary in their department."

### ✅ Answer 6: Subqueries

#### **Solution Using Correlated Subquery:**

```sql
SELECT 
  e.employee_id,
  e.first_name,
  e.last_name,
  e.department_id,
  e.salary
FROM employees e
WHERE e.salary > (
  SELECT AVG(salary)
  FROM employees
  WHERE department_id = e.department_id  -- Correlated!
);
```

#### **Subquery Types:**

**1. Scalar Subquery (Returns single value)**
```sql
-- Compare to overall average
SELECT first_name, salary
FROM employees
WHERE salary > (SELECT AVG(salary) FROM employees);
```

**2. Column Subquery (Returns list of values)**
```sql
-- Find employees in high-revenue departments
SELECT first_name, department_id
FROM employees
WHERE department_id IN (
  SELECT department_id
  FROM departments
  WHERE annual_revenue > 1000000
);
```

**3. Table Subquery (Returns multiple rows/columns)**
```sql
-- Join with subquery
SELECT e.first_name, dept_stats.avg_salary
FROM employees e
JOIN (
  SELECT department_id, AVG(salary) AS avg_salary
  FROM employees
  GROUP BY department_id
) dept_stats ON e.department_id = dept_stats.department_id;
```

**4. Correlated Subquery (References outer query)**
```sql
-- Subquery executes once per outer row
SELECT e1.first_name, e1.salary
FROM employees e1
WHERE salary > (
  SELECT AVG(salary)
  FROM employees e2
  WHERE e2.department_id = e1.department_id  -- References e1
);
```

#### **Subquery Locations:**

**In SELECT (Scalar subquery)**
```sql
SELECT 
  first_name,
  salary,
  (SELECT AVG(salary) FROM employees) AS company_avg,
  salary - (SELECT AVG(salary) FROM employees) AS diff_from_avg
FROM employees;
```

**In FROM (Inline view / derived table)**
```sql
SELECT dept_name, avg_salary
FROM (
  SELECT department_id, AVG(salary) AS avg_salary
  FROM employees
  GROUP BY department_id
) dept_stats
JOIN departments d ON dept_stats.department_id = d.department_id;
```

**In WHERE (Filtering)**
```sql
-- IN subquery
WHERE department_id IN (SELECT department_id FROM departments WHERE location = 'NY')

-- EXISTS subquery
WHERE EXISTS (SELECT 1 FROM orders WHERE orders.customer_id = customers.customer_id)

-- Comparison subquery
WHERE salary > (SELECT AVG(salary) FROM employees)
```

**In HAVING (Filter groups)**
```sql
SELECT department_id, AVG(salary)
FROM employees
GROUP BY department_id
HAVING AVG(salary) > (SELECT AVG(salary) FROM employees);
```

#### **EXISTS vs IN:**

**EXISTS (Usually faster for large datasets)**
```sql
-- Find customers with at least one order
SELECT c.customer_id, c.name
FROM customers c
WHERE EXISTS (
  SELECT 1 
  FROM orders o 
  WHERE o.customer_id = c.customer_id
);

-- Stops at first match (efficient!)
```

**IN (Simpler, good for small lists)**
```sql
-- Find customers with orders
SELECT c.customer_id, c.name
FROM customers c
WHERE c.customer_id IN (
  SELECT customer_id FROM orders
);

-- Builds full list first (can be slow)
```

**NOT EXISTS vs NOT IN:**
```sql
-- ✅ NOT EXISTS: Handles NULLs correctly
WHERE NOT EXISTS (
  SELECT 1 FROM orders WHERE customer_id = customers.customer_id
)

-- ⚠️ NOT IN: Returns no rows if subquery has NULLs!
WHERE customer_id NOT IN (
  SELECT customer_id FROM orders  -- If any NULL, returns empty!
)
```

#### **Common Mistakes:**

**❌ Mistake 1: Multiple-row subquery in scalar context**
```sql
-- Bad: Subquery returns multiple rows
WHERE salary = (SELECT salary FROM employees WHERE department_id = 10);
-- Error: Subquery returns more than 1 row

-- Good: Use IN
WHERE salary IN (SELECT salary FROM employees WHERE department_id = 10);
```

**❌ Mistake 2: Using NOT IN with NULLs**
```sql
-- Bad: Returns no rows if any NULL in subquery
WHERE id NOT IN (SELECT parent_id FROM table);  -- parent_id has NULLs

-- Good: Filter NULLs or use NOT EXISTS
WHERE id NOT IN (SELECT parent_id FROM table WHERE parent_id IS NOT NULL);
-- Or
WHERE NOT EXISTS (SELECT 1 FROM table WHERE parent_id = outer.id);
```

**❌ Mistake 3: Correlated subquery performance**
```sql
-- Bad: Runs subquery for every row (slow!)
SELECT e.name,
  (SELECT COUNT(*) FROM orders WHERE customer_id = e.id) AS order_count
FROM employees e;

-- Good: Use JOIN
SELECT e.name, COUNT(o.order_id) AS order_count
FROM employees e
LEFT JOIN orders o ON e.id = o.customer_id
GROUP BY e.name;
```

### ❓ Question 7: Common Table Expressions (CTEs)

**Interview Question:**
> "Rewrite the previous query using a CTE (WITH clause) to make it more readable."

### ✅ Answer 7: Common Table Expressions (CTEs)

#### **Solution Using CTE:**

```sql
WITH dept_averages AS (
  SELECT 
    department_id,
    AVG(salary) AS avg_salary
  FROM employees
  GROUP BY department_id
)
SELECT 
  e.employee_id,
  e.first_name,
  e.last_name,
  e.department_id,
  e.salary,
  da.avg_salary AS dept_avg,
  e.salary - da.avg_salary AS diff_from_dept_avg
FROM employees e
JOIN dept_averages da ON e.department_id = da.department_id
WHERE e.salary > da.avg_salary;
```

#### **CTE Benefits:**

✅ **Readability:** Break complex queries into logical steps
✅ **Reusability:** Reference CTE multiple times in same query
✅ **Debugging:** Test each CTE separately
✅ **Maintainability:** Easier to modify and understand

#### **Multiple CTEs:**

```sql
WITH 
  -- CTE 1: Department stats
  dept_stats AS (
    SELECT 
      department_id,
      AVG(salary) AS avg_salary,
      COUNT(*) AS employee_count
    FROM employees
    GROUP BY department_id
  ),
  -- CTE 2: High performers
  high_performers AS (
    SELECT 
      e.*,
      ds.avg_salary
    FROM employees e
    JOIN dept_stats ds ON e.department_id = ds.department_id
    WHERE e.salary > ds.avg_salary * 1.2  -- 20% above avg
  )
-- Main query
SELECT 
  hp.first_name,
  hp.salary,
  hp.avg_salary,
  (hp.salary - hp.avg_salary) / hp.avg_salary * 100 AS percent_above_avg
FROM high_performers hp
ORDER BY percent_above_avg DESC;
```

#### **CTE vs Subquery:**

| Aspect | CTE | Subquery |
|--------|-----|----------|
| **Readability** | ✅ High | ⚠️ Can get nested/messy |
| **Reusability** | ✅ Can reference multiple times | ❌ Must repeat code |
| **Debugging** | ✅ Easy to test each step | ⚠️ Harder to isolate |
| **Performance** | Same as subquery | Same as CTE |
| **Recursion** | ✅ Supports recursive CTEs | ❌ No recursion |

**Example: CTE vs Subquery**
```sql
-- With CTE (readable)
WITH high_earners AS (
  SELECT * FROM employees WHERE salary > 80000
)
SELECT department_id, COUNT(*)
FROM high_earners
GROUP BY department_id;

-- With Subquery (harder to read)
SELECT department_id, COUNT(*)
FROM (
  SELECT * FROM employees WHERE salary > 80000
) high_earners
GROUP BY department_id;
```

#### **Recursive CTE (Advanced):**

```sql
-- Employee hierarchy (manager-employee relationships)
WITH RECURSIVE employee_hierarchy AS (
  -- Base case: Top-level managers
  SELECT 
    employee_id,
    first_name,
    manager_id,
    1 AS level
  FROM employees
  WHERE manager_id IS NULL
  
  UNION ALL
  
  -- Recursive case: Employees reporting to previous level
  SELECT 
    e.employee_id,
    e.first_name,
    e.manager_id,
    eh.level + 1
  FROM employees e
  JOIN employee_hierarchy eh ON e.manager_id = eh.employee_id
)
SELECT * FROM employee_hierarchy
ORDER BY level, employee_id;
```

#### **Best Practices:**

**1. Use descriptive CTE names**
```sql
-- Bad
WITH t1 AS (...), t2 AS (...)

-- Good
WITH high_value_customers AS (...),
     recent_orders AS (...)
```

**2. Add comments for complex logic**
```sql
WITH 
  -- Calculate rolling 30-day revenue per customer
  customer_revenue AS (
    ...
  ),
  -- Identify customers in top 10% by revenue
  top_customers AS (
    ...
  )
```

**3. Test CTEs independently**
```sql
-- Test just the CTE

  SELECT department_id, AVG(salary) AS avg_salary
  FROM employees
  GROUP BY department_id
)
SELECT * FROM dept_stats;  -- Verify intermediate results
```

## 📌 Section 5: NULL Handling (3 Questions)

NULL is one of the most misunderstood concepts in SQL. Mastering NULL handling prevents bugs and incorrect results.

### ❓ Question 8: Understanding NULL

**Critical Interview Question:**
> "Explain what NULL represents in SQL and why the following query returns no rows:
> 
> ```sql
> SELECT * FROM employees WHERE email = NULL;
> ```
>
> What's the correct way to find employees with NULL emails?"

### ✅ Answer 8: NULL Behavior

#### **What is NULL?**

NULL represents **unknown** or **missing** data. It's NOT:
- ❌ Not zero (0)
- ❌ Not empty string ('')
- ❌ Not FALSE

NULL means **"I don't know what the value is"**

#### **Why `WHERE email = NULL` Returns No Rows:**

```sql
-- This ALWAYS returns no rows
SELECT * FROM employees WHERE email = NULL;

-- Why? NULL = NULL evaluates to NULL (not TRUE!)
-- NULL is unknown, so comparing unknowns is also unknown
```

**NULL Comparison Results:**
```sql
NULL = NULL    → NULL (not TRUE!)
NULL != NULL   → NULL (not TRUE!)
NULL > 5       → NULL
NULL < 100     → NULL
```

#### **Correct Way to Check for NULL:**

```sql
-- ✅ Correct: Use IS NULL
SELECT * FROM employees WHERE email IS NULL;

-- ✅ Correct: Use IS NOT NULL
SELECT * FROM employees WHERE email IS NOT NULL;

-- ❌ Wrong
WHERE email = NULL    -- Always FALSE
WHERE email != NULL   -- Always FALSE
```

#### **NULL in Arithmetic:**

```sql
-- Any operation with NULL returns NULL
SELECT 
  10 + NULL AS result1,     -- NULL
  100 * NULL AS result2,    -- NULL
  NULL / 5 AS result3,      -- NULL
  CONCAT('Hello', NULL) AS result4;  -- NULL (in most databases)
```

**Real-world impact:**
```sql
-- Calculate total compensation (salary + bonus)
SELECT 
  first_name,
  salary,
  bonus,
  salary + bonus AS total_comp  -- NULL if bonus is NULL!
FROM employees;

-- ✅ Fix: Use COALESCE
SELECT 
  first_name,
  salary,
  bonus,
  salary + COALESCE(bonus, 0) AS total_comp  -- Treat NULL as 0
FROM employees;
```

#### **NULL in Aggregate Functions:**

```sql
-- Aggregate functions IGNORE NULL values
SELECT 
  COUNT(*) AS total_rows,           -- 5 (includes NULLs)
  COUNT(email) AS non_null_emails,  -- 4 (excludes NULLs)
  AVG(bonus) AS avg_bonus           -- Average of non-NULL bonuses only
FROM employees;
```

#### **NULL in WHERE Clause Logic:**

```sql
-- AND with NULL
TRUE AND NULL  → NULL
FALSE AND NULL → FALSE
NULL AND NULL  → NULL

-- OR with NULL
TRUE OR NULL   → TRUE
FALSE OR NULL  → NULL
NULL OR NULL   → NULL

-- NOT with NULL
NOT NULL → NULL
```

**Example:**
```sql
-- Find employees with salary > 50K OR unknown department
SELECT * FROM employees
WHERE salary > 50000 OR department_id IS NULL;

-- Find employees with salary > 50K AND department assigned
SELECT * FROM employees
WHERE salary > 50000 AND department_id IS NOT NULL;
```

#### **NULL in ORDER BY:**

```sql
-- NULL sorting behavior (database-dependent)
ORDER BY email  -- NULLs first or last?

-- Explicit control
ORDER BY email NULLS FIRST   -- NULLs at beginning
ORDER BY email NULLS LAST    -- NULLs at end
```

#### **NULL-Safe Functions:**

**COALESCE (Returns first non-NULL value)**
```sql
SELECT 
  COALESCE(email, phone, 'No Contact') AS contact,
  COALESCE(bonus, 0) AS bonus_amount
FROM employees;
```

**NULLIF (Returns NULL if equal)**
```sql
-- Avoid division by zero
SELECT 
  revenue / NULLIF(customer_count, 0) AS revenue_per_customer
FROM sales;
-- If customer_count = 0, NULLIF returns NULL, result is NULL
```

**IFNULL / NVL (Database-specific)**
```sql
-- MySQL/SQLite
IFNULL(bonus, 0)

-- Oracle
NVL(bonus, 0)

-- Standard SQL (most portable)
COALESCE(bonus, 0)
```

#### **NULL in CASE Expressions:**

```sql
SELECT 
  first_name,
  CASE 
    WHEN email IS NULL THEN 'No Email'
    WHEN email LIKE '%@company.com' THEN 'Company Email'
    ELSE 'External Email'
  END AS email_type
FROM employees;
```

#### **Common NULL Pitfalls:**

**Pitfall 1: NOT IN with NULLs**
```sql
-- ❌ Returns no rows if subquery has any NULL!
SELECT * FROM employees
WHERE department_id NOT IN (SELECT department_id FROM departments);
-- If any department_id is NULL, entire result is empty!

-- ✅ Fix: Filter NULLs or use NOT EXISTS
WHERE department_id NOT IN (
  SELECT department_id FROM departments WHERE department_id IS NOT NULL
);
```

**Pitfall 2: Counting NULLs**
```sql
-- ❌ Wrong: COUNT(*) includes NULLs
SELECT COUNT(*) FROM employees WHERE email IS NULL;  -- Counts rows

-- ✅ Correct: Use SUM with CASE
SELECT SUM(CASE WHEN email IS NULL THEN 1 ELSE 0 END) AS null_count
FROM employees;
```

**Pitfall 3: String concatenation**
```sql
-- ❌ NULL makes entire result NULL
SELECT first_name || ' ' || middle_name || ' ' || last_name
FROM employees;
-- If middle_name is NULL, entire name is NULL!

-- ✅ Fix: Use COALESCE
SELECT first_name || ' ' || COALESCE(middle_name || ' ', '') || last_name
FROM employees;
```

#### **Interview Answer Template:**

1. **Define NULL:** "NULL represents unknown or missing data"
2. **Explain comparison:** "NULL = NULL returns NULL (unknown), not TRUE"
3. **Show correct syntax:** "Use IS NULL / IS NOT NULL, not = NULL"
4. **Mention impact:** "Aggregate functions ignore NULLs, arithmetic returns NULL"
5. **Best practice:** "Always handle NULLs explicitly with COALESCE or IS NULL checks"

````python
%sql
-- Demo: NULL Handling

-- Update one employee to have NULL department
UPDATE workspace.default.employees
SET department_id = NULL
WHERE employee_id = 6;

-- Wrong: Returns no rows
SELECT * FROM workspace.default.employees
WHERE department_id = NULL;

-- Correct: Find employees with NULL department
SELECT 
  employee_id,
  CONCAT(first_name, ' ', last_name) AS full_name,
  department_id
FROM workspace.default.employees
WHERE department_id IS NULL;

-- COALESCE: Replace NULL with default
SELECT 
  employee_id,
  CONCAT(first_name, ' ', last_name) AS full_name,
  COALESCE(department_id, 0) AS dept_id,
  salary,
  COALESCE(salary * 0.1, 0) AS bonus
FROM workspace.default.employees;

-- NULL in aggregations
SELECT 
  COUNT(*) AS total_employees,
  COUNT(department_id) AS employees_with_dept,
  COUNT(*) - COUNT(department_id) AS employees_without_dept
FROM workspace.default.employees;
````

### ❓ Question 10: Date Manipulation

### ✅ Answer 10: Date Functions

#### **Solutions:**

```sql

FROM employees;

SELECT first_name, last_name, hire_date
FROM employees
WHERE DAYOFWEEK(hire_date) = 2;  -- Monday (1=Sunday, 2=Monday, ...)
-- Or
WHERE DATE_FORMAT(hire_date, 'EEEE') = 'Monday';
```

#### **INSERT Variants:**

**Insert with explicit columns (recommended)**
```sql
INSERT INTO employees (employee_id, first_name, last_name, salary)
```

**Insert all columns (risky - order-dependent)**

**Insert from SELECT (copy data)**

#### **UPDATE Patterns:**

**Simple UPDATE**
```sql
UPDATE employees
SET salary = 80000
WHERE employee_id = 1;
```

**Update multiple columns**
```sql
UPDATE employees
SET 
  salary = salary * 1.10,
  last_updated = CURRENT_TIMESTAMP
```

**Update from another table (JOIN)**
```sql
-- Give bonus based on department performance
UPDATE employees e
SET e.bonus = e.salary * d.bonus_percentage
FROM departments d
WHERE e.department_id = d.department_id
  AND d.performance_rating = 'Excellent';
```

**Update with CASE**
```sql
UPDATE employees
SET salary = CASE
  WHEN department_id = 1 THEN salary * 1.15  -- 15% for Engineering
  WHEN department_id = 2 THEN salary * 1.10  -- 10% for Sales
  ELSE salary * 1.05                         -- 5% for others

```

#### **DELETE Patterns:**

**Delete with WHERE**
```sql

WHERE hire_date < '2020-01-01';
```

**Delete all rows (use with caution!)**

**Delete with subquery**
```sql

WHERE employee_id IN (
```

#### **Transactions (ACID)**

#### **Common Mistakes:**

**❌ Mistake 1: Forgetting WHERE clause**
```sql
-- Bad: Updates ALL rows!
UPDATE employees SET salary = 100000;  -- Everyone gets $100K!

-- Good: Always use WHERE
UPDATE employees SET salary = 100000 WHERE employee_id = 1;
```

**❌ Mistake 2: Not testing with SELECT first**
```sql
-- ✅ Best practice: Test query first
SELECT * FROM employees WHERE hire_date < '2020-01-01';
-- Verify results, THEN delete
DELETE FROM employees WHERE hire_date < '2020-01-01';
```

**❌ Mistake 3: No backup before bulk changes**
```sql
-- Before large UPDATE/DELETE:
-- 1. Create backup
CREATE TABLE employees_backup AS SELECT * FROM employees;
-- 2. Make changes
DELETE FROM employees WHERE ...;
-- 3. Verify
-- 4. Drop backup
DROP TABLE employees_backup;
```

### 🚀 Key Takeaways:

**Query Execution Order (Memorize this!):**
```
FROM → WHERE → GROUP BY → HAVING → SELECT → ORDER BY → LIMIT
```

**NULL Rules:**
- NULL = NULL returns NULL (not TRUE!)
- Use IS NULL / IS NOT NULL
- Aggregates ignore NULLs
- Use COALESCE for defaults

**JOIN Selection:**
- INNER JOIN: Only matching rows
- LEFT JOIN: All left table rows + matches
- RIGHT JOIN: All right table rows + matches
- FULL OUTER: All rows from both tables

**WHERE vs HAVING:**
- WHERE: Filters rows before GROUP BY
- HAVING: Filters groups after GROUP BY

**Performance Tips:**
- Select only needed columns (not SELECT *)
- Add WHERE clause before GROUP BY
- Use indexes on JOIN and WHERE columns
- Test with SELECT before UPDATE/DELETE
